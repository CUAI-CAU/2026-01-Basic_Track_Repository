# 회귀
## 1.회귀: 여러 개의 독립변수와 한 개의 종속변수 간의 상관관계를 모델링 하는 기법 통칭
#### ex)아파트 방 개수, 방 크기, 주변 학군 등 여러 개의 변수에 따라 아파트 가격이라는 종석변수가 어떤 관계 나타내는 지 모델링하고 예측
#### ->회귀는 회 계수의 선형/비선형, 독립변수의 개수, 종속 변수의 개수에 따라 여러 유형으로 나누어짐


## 회귀 계수(Regression coefficients):독립 변수 값에 영향을 미친다.
### ->회귀 계수가 회귀에서 가장 중요하고 이것이 선형인지 아니냐 에 따라 선형 회귀와 비선형 회귀로 나누어짐
### ->머신 러닝 관점에서 보면 독립 변수는 피처 종속 변수는 결정 값으로 피처와 결정 값 데이터 기반에서 학습 통해 최적의 회귀 계수를 찾는 것이다.

### 회귀 분석: 데이터 값이 평균과 같은 일정한 값으로 돌아가려는 경향을 이용한 통계 기법



### 회귀 분류

### : 실제 값과 예측값의 차이(오류의 제곱 값)를 최소화 하는 직선형 회귀선을 최적화하는 방식

#### 대표적 선형 회귀 모델: 일반 선형 회귀, 릿지, 라쏘, 엘라스틱넷, 로지스틱 회귀

#### ->교재에서 실제 주택 가격은 1차 함수 값에서 실제 값 만큼의 오류 값을 뺀(또는 더한) 값이 된다.



### 잔차 : 회귀 모델의 차이에 따른 오류 값(남은 오류)->따라서 최적의 회귀 모델 만드는 것은 잔차(오류 값)의 합이 최소가 되는 모델을 만든다는 것이고 오류 값 합이 최소가 되는 회기 계수를 찾는 것이다

#### =>일반적으로 RSS를 많이 사용한다.(미분 등 계산 편리 위해)=>이 RSS를 최소로 하기 위한 회귀 계수를 머신러닝을 통해 찾는다.
#### =>비용함수(=손실 함수)로서 표현하기도 함




## 3.비용 최소화 하기-경사 하강법(Gradient Descent) 소개

### :비용함수 최소가 되는 W 파라미터 구하는 방법 ->경사 하강법
### 경사 하강법->고차원 방정식의 어려움도 해결할 수 있고 RSS 최소화 하는 방법도 직관적으로 제공하는 방식이다.
### ->데이터를 기반으로 학습한다라는 개념을 가능하게 만든 기법

### 경사 하강법: 점진적으로 반복적 계산을 통해 W 파라미터 값을 업데이트 하면서 오류 값이 최소가 되는 W 파라미터를 구하는 방식

#### ->경사 하강법은 예측값과 실제 값이 작아지는 방향성을 가지고 W 파라미터를 지속적 보정을 한다. ->더 이상 오류 갓이 작아지지 않으면 그 오류 값을 소 비용으로 보고 W값을 최적 파라미터로 반환 ->미분을 적용하다가 1차 함수 기울기가 감소하지 않는 지점을 비용함수 최소 지 간주하고 w반환


### -> RSS는 w0, w1(회귀 계수)에 대한 함수이다.
### -> x(독립변수)와 y(종속변수)는 주어진 데이터이므로 고정된 값으로 간주한다.

$$
J(w_0, w_1) = \frac{1}{n} \sum_{i=1}^{n} \left( y_i - (w_0 + w_1 x_i) \right)^2
$$

#### - y는 실제값이고, (w0 + w1 x)는 예측값이다.
#### - 두 값의 차이는 오차(error)를 의미한다.
#### - 이 오차를 제곱하여 모두 더한 값이 RSS이며, 이를 최소화하는 w0, w1을 찾는 것이 목표이다.


### 선형 회귀 비용함수의 편미분(연쇄 법칙에 따른)

비용함수:

$$
R(w) = \frac{1}{N} \sum_{i=1}^{N} \left(y_i - (w_0 + w_1 x_i)\right)^2
$$

$w_1$에 대한 편미분:

$$
\frac{\partial R(w)}{\partial w_1}
= -\frac{2}{N} \sum_{i=1}^{N} x_i \left(y_i - (w_0 + w_1 x_i)\right)
$$

$w_0$에 대한 편미분:

$$
\frac{\partial R(w)}{\partial w_0}
= -\frac{2}{N} \sum_{i=1}^{N} \left(y_i - (w_0 + w_1 x_i)\right)
$$

### 이런 식으로 계속 w1,w0을 업데이트 하면서 R(w)가 최소가 되는 w1,w0 값을 구할 수 있다.
#### ->업데이트는 새로운 w1을 이전 w1 편미분 결과값을 마이너스 하면서 적용
#### ex) 이전 w1- 새로운 w1

### 학습률: 편미분 값이 너무 클 수 있기에 보정 계수법을 하는 것


### [경사하강법 단계]

- **Step 1**  
  $w_1, w_0$를 임의의 값으로 설정하고 첫 비용 함수 값을 계산합니다.

- **Step 2**  
  $w_1$을 다음과 같이 업데이트합니다:

$$
w_1 = w_1 + \eta \frac{2}{N} \sum_{i=1}^{N} x_i (실제값_i - 예측값_i)
$$

$$
w_0 = w_0 + \eta \frac{2}{N} \sum_{i=1}^{N} (실제값_i - 예측값_i)
$$

업데이트한 후 비용 함수 값을 다시 계산합니다.

- **Step 3**  
  비용 함수가 감소하는 방향으로 주어진 횟수만큼 Step 2를 반복하면서  
  $w_1$과 $w_0$를 계속 업데이트합니다.

## get_weight_updates() 
### ->함수-입력 배열 X값에 대한 예측 배열 y_pred는 np.dot(X,w1.T)+W0으로 구함

### w0,w1초기화 후 iters 개수만큼 반복하면서 w1,w0 업데이트 
#### ->100개의 데이터가 있다면 예측 값은 w0+X(1)*w1+X(2)*w0+...+X(100)*w1이고 이는 입력 배열 X와 w1 배열의 내적과 동일 ->따라서 넘파이의 내적 여산이 dot을 통해 예측 배열 값 계산

#### ->비용함수를 미분해서 미분 함수의 최솟값을 구해야 하는데 두 개의 파라미터가 있기 때문에 각각 편미분(두 개를 따로 미분한다는 뜻)을 해야 한다.



```python
#w1과 w0을 업데이트 할 w1_update,w2_update를 반환
def get_weight_updates(w1,w0,X,y,learning_rate=0.01): #틀린 만큼 조금씩 개선해주는 함수
    N=len(y)
    #먼저 w1_update,w0_update를 각각 w1,w0의 shpae와 동일한 크기를 가진 0값으로 초기화
    w1_update=np.zeros_like(w1) 
    w0_update=np.zeros_like(w0)
    #예측 배열 계산하고 예측과 실제 값의 차이를 계산
    y_pred=np.dot(X,w1.T)+w0 #np.dot 각원소끼리 곱하고 더하는 함수 #.T로 전치를 하는 거는 행열 바꾸기 위한 거임
    diff=y-y_pred #얼마나 틀렸는지 확인

    #w0_update를 dot 행렬 연산으로 구하기 위해 모두 1값을 가진 행렬 생성
    w0_factors=np.ones((N,1))
    #w1과 w0을 업데이트할 w1_update와 w0_update 계산
    w1_update=-(2/N)*learning_rate*(np.dot(X.T,diff))
    w0_update=-(2/N)*learning_rate*(np.dot(w0_factors.T,diff))
    return w1_update,w0_update
```

### gradient_descent_steps()함수 :get_weight_updates()를 경사하강 방식으로 반복 수행헤 w0,w1 업데이트 하는 함수

```python
#입력 인자 iters로 주어진 횟수만큼 반복적으로 w1,w0을 업데이트 적용을 함
def gradient_descent_steps(X,y,iters=10000):
    #w0과 w1을 모두 0으로 초기화
    w0=np.zeros((1,1))
    w1=np.zeros((1,1))
    #인자로 주어진 iters 만큼 반복적으로 get_weight_updates() 호출해 w1,w0 업데이트 행
    for ind in range(iters):
        w1_update,w0_update=get_weight_updates(w1,w0,X,y,learning_rate=0.01) 
        w1=w1-w1_update
        w0=w0-w0_update
    return w1,w0
```

### gradient_descent_steps() 호출해 w1,w0 구하고 최종적으로 RSS 차이를 계산하는 get_cost() 함수 생성하고 이를 이용해 경사 하강법의 예측 오류도 계산

```python
def get_cost(y,y_pred):
    N=len(y)
    cost=np.sum(np.square(y-y_pred))/N
    return cost
w1,w0=gradient_descent_steps(X,y,iters=1000)
print('w1:{0:.3f} w0:{1:.3f}'.format(w1[0,0],w0[0,0]))
y_pred=w1[0,0]*X+w0
print('Gradient Descent Total Cost:{0:.4f}'.format(get_cost(y,y_pred)))
```

## 확률적 경사하강법 (stochastic gradient descent)
#### : 전체 입력 데이터로 w가 업데이트되는 값 계산하는 것이 아닌 일부 데이터만 이용해 W가 업데이트 되는 값 계산해 속도가 빠름
#### 경사 하강법(시간이 오래 걸림)->대규모 대용량 데이터의 경우 대부분 확률적 경사 하강법 사용

## (미니배치)확률적 경사 하강법 구현 -stochastic_gradient_descent_steps()함수로 구현)
#### ->X,y데이터에서 batch size만큼 데이터 추출해 이를 기반 w1_update, w0_update 계산 부분만 차이가 있다
#### ->np.random.permutation(X.shpae[0]) 행 인덱스들을 섞은 넘파이 array만들고 이후 X의 슬라이싱과 Y슬라이싱을 통해 batch_size만큼 빼낸다. 나머지는 동일

```python
def stochastic_gradient_descent_steps(X,y,batch_size=10,iters=1000):
    w0=np.zeros((1,1))
    w1=np.zeros((1,1))
    for ind in range(iters):
        np.random.seed(ind)
        #전체 X,y 데이터에서 랜덤하게 batch_size만큼 데이터를 추출한 sample_X,sample_y로 저장
        stochastic_random_index=np.random.permutation(X.shape[0]) # 행의 수 안에서 무작위로 섞은 배열 반환
        sample_X=X[stochastic_random_index[0:batch_size]]
        sample_y=y[stochastic_random_index[0:batch_size]]
        #랜덤하게 batch_size만큼 추출된 데이터를 기반으로 w1_update,w0_update 계산후 업데이트
        w1_update,w0_update=get_weight_updates(w1,w0,sample_X,sample_y,learning_rate=0.01) 
        w1=w1-w1_update
        w0=w0-w0_update
    return w1,w0
```

```python
w1,w0=stochastic_gradient_descent_steps(X,y,iters=1000)
print('w1:{0:.3f} w0:{1:.3f}'.format(w1[0,0],w0[0,0]))
y_pred=w1[0,0]*X+w0
print('Gradient Descent Total Cost:{0:.4f}'.format(get_cost(y,y_pred)))
```

## 회귀 평가 지표
#### 실제 값과 회귀 예측값 차이를 기반으로 한 지표가 중심 ->그냥 더할 경우 +,-가 섞여 오류가 상쇄된다.
#### ->따라서 오류의 절댓값 평균이나 제곱 또는 제곱한 뒤 루트 씌운 평균 값을 구함



| 평가 지표 | 설명 | 수식 |
| :--- | :--- | :--- |
| **MAE** | **Mean Absolute Error**(MAE)이며 실제 값과 예측값의 차이를 절댓값으로 변환해 평균한 것입니다. | $$MAE = \frac{1}{n} \sum_{i=1}^{n} |Y_i - \hat{Y}_i|$$ |
| **MSE** | **Mean Squared Error**(MSE)이며 실제 값과 예측값의 차이를 제곱해 평균한 것입니다. | $$MSE = \frac{1}{n} \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2$$ |
| **RMSE** | MSE 값은 오류의 제곱을 구하므로 실제 오류 평균보다 더 커지는 특성이 있으므로 **MSE에 루트를 씌운 것**이 RMSE(Root Mean Squared Error)입니다. | $$RMSE = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2}$$ |
| **$R^2$** | 분산 기반으로 예측 성능을 평가합니다. 실제 값의 분산 대비 예측값의 분산 비율을 지표로 하며, **1에 가까울수록 예측 정확도가 높습니다.** | $$R^2 = \frac{\text{예측값 Variance}}{\text{실제값 Variance}}$$ |

#### RMSE의 경우 사이킷런에서 제공하지 않기에 MSE에 제곱근을 씌워서 개선하는 함수를 만들어야 한다.
| 평가 방법 | 사이킷런 평가 지표 API | Scoring 함수 적용 값 |
|-----------|----------------------|----------------------|
| MAE | metrics.mean_absolute_error | 'neg_mean_absolute_error' |
| MSE | metrics.mean_squared_error | 'neg_mean_squared_error' |
| RMSE | metrics.mean_squared_error (squared=False 설정) | 'neg_root_mean_squared_error' |
| MSLE | metrics.mean_squared_log_error | 'neg_mean_squared_log_error' |
| R² | metrics.r2_score | 'r2' |

#### ->MAE score 파라미터에 neg라고 써있는 것은 음수를 가진다는 의미가 Scoring 함수가 큰 값을 가지면 좋은 평가로 인식하기 때문에 음수를 곱해서 하는 것이다. neg가 없다면 정상적으로는 양수 값을 반환한다.

## 학습 데이터와 테스트 데이터 분리후 예측 ->이후 mean_squared_error()와 r2_score() API를 이용해 MSE와 R2_score 측정

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
y_target=bostonDF['PRICE']
X_data=bostonDF.drop(['PRICE'],axis=1,inplace=False)
X_train,X_test,y_train,y_test=train_test_split(X_data,y_target,test_size=0.3,random_state=156)

#선형 회귀 OLS로 학습/예측/평가 수행
lr=LinearRegression()
lr.fit(X_train,y_train)
y_preds=lr.predict(X_test)
mse=mean_squared_error(y_test,y_preds)
rmse=np.sqrt(mse)
print('MSE: {0:.3f},RMSE: {1:.3f}'.format(mse,rmse))
print('Variance score: {0:.3f}'.format(r2_score(y_test,y_preds)))
```

## 5개의 폴드 세트에서 cross_val_score을 통해 MSE와 RMSE 구하기
#### 앞에서 설명한 표에서 봤듯이 scoring='neg_mean_squared_error'를 넣어야 한다.

## 다항 회귀와 과(대)적합/과소적합 이해
### 다항(Polynomial) 회귀
#### :회귀가 독립변수의 단항식이 아닌 2,3차 방정식과 같은 당항식으로 표현되는 것
#### 주의점: 다항회귀는 선형회귀이다. / 회귀의 선형 또는 비선형을 나누는 것은 회귀 계수가 선형 비선형에 영향을 받지 독립변수의 선형/비선형과는 무관하다

#### -앞에서는 일차 방정식 형태로 회귀를 표현했으나 현실의 모든 것을 직선으로만 표현할 수 없기에 단항식이 아니라 다항식으로 표현하는 것 필요
 

### 사이킷런은 다항회귀 위한 클래스를 명시적으로 제공하지 않기에 비선형 함수를 선형 모델에 적용시키는 방식 통해 구현
#### 구현방법: (1)PolynomialFeatures 클래스를 통해 피처를 Polynaomial(다항식) 피처로 변환 / fit() & transform() 매서드로 수행
#### ->단항값[x1,x2]를 2차 다항값으로[1,x1,x2,x1^2,x1*x2,x2^2]로 변환하는 예제

```python
from sklearn.preprocessing import PolynomialFeatures
import numpy as np

#다항식으로 변환한 단항식 생성 [[0,1],[2,3]]의 2X2 행렬 생성
X=np.arange(4).reshape(2,2)
print('일차 단항식 계수 피처:\n',X)
#degree=2인 2차 다항식으로 변환하기 위해 PolynomialFeatures를 이용해 변환
poly=PolynomialFeatures(degree=2)
poly.fit(X)
poly_ftr=poly.transform(X)
print('변환된 2차 다항식 계수 피처:\n',poly_ftr)

```

## 3차 다항함수 회귀 함수식을  PolynomialFeatures와 LinearRegression 클래스 이용해 우도
#### ->3차 다항회귀 함수 임의 설정후 회귀계수 예측
#### 임의로 정한 회귀의 결정 함수식: $$ y = 1 + 2x_1 + 3x_1^2 + 4x_2^3 $$ ->이를 위한 polynomial_func()를 만든다. ->이 함수는 다항 계수 피처 값 입력되면 결정값 반환


```python
def polynomial_func(X):
    y=1+2*X[:,0]+3*X[:,0]**2+4*X[:,1]**3
    return y
X=np.arange(4).reshape(2,2)
print('일차 단항식 계수 feature: \n',X)
y=polynomial_func(X)
print('삼차 다항식 결정값:\n',y)
```

#### 이제 일차 단항식 계수를 삼차 다항식 계수로 변환하고 선형회귀에 적용하면 다항회귀로 구현 된다.

```python
#3차 다항식 변환
poly_ftr=PolynomialFeatures(degree=3).fit_transform(X) #기존: [x1, x2]확장: [1, x1, x2, x1², x1x2, x2², x1³, x1²x2, x1x2², x2³]
print('다항식 계수 feature \n',poly_ftr)

#Linear Regression에 3차 다항식 계수 feature와 3차 다항식 결정값으로 학습 후 회귀 계수 확인
model=LinearRegression()
model.fit(poly_ftr,y) #선형회귀로 학습
print('Polynomial 회귀 계수\n',np.round(model.coef_,2))
print('Polynomial 회귀 계수 shape:',model.coef_.shape)

```

#### 앞의 방식을 보면 일차 단항식 계수 피처는 2개였지만 3차 다항식 Polynomial 변환 이후에는 다항식 계수 피처가 10개로 늘어남
#### ->이 피처 데이터 세트에 LinearRegression을 통해 3차 다항회귀를 적용하면 회귀 계수가 10개로 늘어남
#### 결과: 원래 계수인 [1,2,0,3,0,0,0,0,0,4] 차이는 있지만 다항회귀로 근사하고 있다.

#### ->따라서 위의 방식대로 사이킷런은 PolynomialFeatures로 피처 변환후 LinearRegression 클래스로 다항회귀 구현



## 앞에서와 달리 피처 변환과 선형 회귀 적용 별도가 아닌 pipline 객체를 이용해 한번에 구현방법

## 다항 회귀를 이용한 과소적합 및 과적합 이해

### 과적합:차수가 높을 수록 학습 데이터에만 너무 맞춰지게 되어 테스트 데이터 환경에서 예측도가 떨어짐
#### :다항회귀의 피처는 직선적 관계가 아닌 복잡한 다항 관계 모델링 가능(차수가 높아질수록 복잡한 피처간 관계 가능)
#### ->그러나 degree(차수)가 높일 수록 학습데이터에만 너무 맞춰지게 되어 테스트 데이터 환경에서 예측도가 틀어짐 (과적합)


#### 실행예제-원래 데이터 세트는 피처 X와 target y가 포함된 코사인 그래프 관계 가지도록 함 ->이에 기반 다항 회귀 차수 변화시키며 회귀 예측 곡선 과 예측 정확도 비교하는 것

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score
%matplotlib inline

#임의의 값으로 구성된 X값에 대해 코사인 변환 값 변환
def true_fun(x):
    return np.cos(1.5*np.pi*x)
#X는 0부터 1까지 30개의 임의의 값을 순서대로 샘플링한 데이터
np.random.seed(0)
n_samples=30
X=np.sort(np.random.rand(n_samples))

#y값은 코사인 기반의 true_fun()에서 약간의 나이즈 값을 더한 값
y=true_fun(X)+np.random.randn(n_samples)*0.1



```

#### 예측 결과를 비교할 다항식 차수를 1,4,15로 변경하면서 예측 결과 비교 (cross_val_score()로 MSE 구해 차수별 예측 성능 비교)

```python
plt.figure(figsize=(14,5))
degrees=[1,4,15]
#다항 회귀의 차수(degree)를 1,4,15로 각각 변화시키면서 비교
for i in range(len(degrees)):
    ax=plt.subplot(1,len(degrees),i+1)
    plt.setp(ax,xticks=(),yticks=()) #속성 관련 함수,x,y 둘다 눈금 없게[
    #개별 degree 별로 Polynomial 변환
    polynomial_features=PolynomialFeatures(degree=degrees[i],include_bias=False) #include_bias는 상수항을 포함관련 파라미터
    linear_regression=LinearRegression()
    pipeline=Pipeline([('polynomial_features',polynomial_features),('linear_regression',linear_regression)])
    pipeline.fit(X.reshape(-1,1),y)
    #교차 검증으로 다항 회귀 평가
    scores=cross_val_score(pipeline,X.reshape(-1,1),y,scoring='neg_mean_squared_error',cv=10)
    #Pipeline 구성하는 세부 객체를 접근하는 named_steps['객체명']을 이용해 회귀계수 추출
    cofficients=pipeline.named_steps['linear_regression'].coef_
    print('\nDegree {0} 회귀 계수는 {1}입니다.'.format(degrees[i],np.round(cofficients,2)))
    print('\nDegree {0} MSE는 {1}입니다.'.format(degrees[i],np.round(-1*np.mean(scores),2)))
    #0부터 1까지 테스트 데이터 세트를 100개로 나눠 예측 수행
    #테스트 데이터 세트에 휘귀 예측 수행하고 예측 곡선과 실제 곡선 그려서 비교
    X_test=np.linspace(0,1,100) #0부터 1까지를 100개로 나눈 것 생성
    #예측값 곡선
    plt.plot(X_test,pipeline.predict(X_test[:,np.newaxis]),label='Model')
    #실제 값 곡선
    plt.plot(X_test,true_fun(X_test),'--',label='True Function')
    plt.scatter(X,y,edgecolor='b',s=20,label='Samples')
    plt.xlabel('x'); plt.ylabel('y'); plt.xlim((0,1));plt.ylim((-2,2));plt.legend(loc='best')
    plt.title('Degree {}\nMSE={:.2e}(+/- {:.2e}'.format(degrees[i],-scores.mean(),scores.std()))
plt.show()









```

#### ->맨 왼쪽 1차의 경우 단순 선형과 같음 MSE가 0.41이고 예측했다기에는 한계가 있음 (과소적합)
#### ->가운데는 실제 데이터 세트와 유사하고 MSE도 0.04로 뛰어나 예측을 한다. (적절)
#### ->맨 오른쪽은 MSE가 너무 높다. 학습 데이터만 잘 반영하고 테스트 데이터는 반영을 잘 못하는 과적합 발생 (과적합)

#### ->일반적으로 변향과 분산은한쪽이 높으면 한쪽이 낮아짐
#### ex) 편향 높으면 분산 낮아짐(과소적합), 분산 높으면 편향이 낮아짐(과적합)
#### ->편향이 높으면 전체 오류가 높고 편향을 낮추면 분산이 높아지고 전체 오류도 낮아짐, 편향 낮추고 분산 높이면 전체 오류가 낮아지는 골디락스 지점 통과하고 계속 높이면 오히려 예측성능 다시 낮아짐
