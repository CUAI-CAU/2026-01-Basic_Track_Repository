### 1. 분류(Classification)

<aside>
    
**용어 정리**
- 지도 학습(Supervised Learning) : 학습 데이터로 학습한 후 테스트 데이터 세트에서 예측
- 피처(Feature) : 모델에 입력되는 데이터의 속성(ex. 키, 몸무게 …)
    - 의미 있는 데이터를 학습해야 함 → 입력 제어
- 레이블(Lable) / 타겟(Target) : 모델이 예측해야하는 정답(ex. 남자/여자 …)
    - 제대로 된 정답을 학습해야 함 → 출력 제어
- 하이퍼 파라미터 : 학습 전 사람이 직접 설정해 모델 구조와 학습 방식을 제어하는 값(ex. KNN의 k값, 결정트리의 max_depth …)
    - 모델 성능 최적화의 핵심 → 학습 방식 제어
    
    cf. 파라미터 : 모델 내부에서 학습을 통해 자동으로 결정되는 값(ex. 가중치 …)
  
</aside>    

**ML 분류 예측 프로세스** : 데이터 세트 분리 → 모델 학습 → 예측 수행 → 평가

**sklearn 모듈**

- sklearn.datasets : 사이킷런에서 자체적으로 제공하는 데이터 세트를 생성하는 모듈
- sklearn.tree : 트리 기반 ML 알고리즘을 구현한 모듈
- sklearn.model_selection : 데이터 분리 및 최적의 하이퍼 파라미터로 평가하기 위한 모듈
- sklearn.metrics : 성능을 측정하는 평가 모듈

**관련 sklearn API**

- train_test_split( dataset, labels, test_size=0.n, random_state=n ) : 학습 데이터와 테스트 데이터를 test_size의 비율로 분할하는 함수
    - test_size : 전체 데이터 세트 중 테스트 데이터 세트의 비율
    - random_state : 호출할 때마다 같은 학습/테스트 데이터 세트를 생성하기 위한 난수
    - X_train, X_test, y_train, y_test 생성함
- DecisionTreeClassifier(  ) : 의사 결정 트리 클래스 객체 ; random_state 사용 가능
- Instance.fit( X_train, y_train ) : 학습용 데이터 세트로 학습을 수행하는 메서드
- Instance.predict( X_test ) : 테스트 데이터 세트를 입력하면 예측 값을 반환하는 메서드
- accuracy_score( y_test, predict ) : 예측 데이터 세트가 실제 레이블과 얼마나 맞는 지 평가하는 함수

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
```

```python
import pandas as pd

# 붓꽃 데이터 세트를 로딩합니다. 
iris = load_iris()

# iris.data는 Iris 데이터 세트에서 피처(feature)만으로 된 데이터를 numpy로 가지고 있습니다. 
iris_data = iris.data

# iris.target은 붓꽃 데이터 세트에서 레이블(결정 값) 데이터를 numpy로 가지고 있습니다. 
iris_label = iris.target
print('iris target값:', iris_label)
print('iris target명:', iris.target_names)

# 붓꽃 데이터 세트를 자세히 보기 위해 DataFrame으로 변환합니다. 
iris_df = pd.DataFrame(data=iris_data, columns=iris.feature_names)
iris_df['label'] = iris.target
iris_df.head(3)
```

```python
X_train, X_test, y_train, y_test = train_test_split(iris_data, iris_label, 
                                                    test_size=0.2, random_state=11)
```

```python
# DecisionTreeClassifier 객체 생성 
dt_clf = DecisionTreeClassifier(random_state=11)

# 학습 수행 
dt_clf.fit(X_train, y_train)
```

```python
# 학습이 완료된 DecisionTreeClassifier 객체에서 테스트 데이터 세트로 예측 수행. 
pred = dt_clf.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print('예측 정확도: {0:.4f}'.format(accuracy_score(y_test,pred)))
```
