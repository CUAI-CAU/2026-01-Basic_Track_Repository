```python
from sklearn.datasets import load_iris

iris_data = load_iris()
print(type(iris_data))

# bunch클래스는  파이썬 딕셔너리 자료형과 유사.
# 데이터 세트에 내장돼 있는 대부분의 데이터 세트는 이와 가티 딕셔너리 형태의 값 반환.
```

```python
# 데이터 키는 피처들의 데이터 값을 가리킨다. 
# 데이터 세트가 딕셔너리 형태이기 때문에 피처 데이터 값을 추출하기 위해서는 데이터 세트.data(또는 데이터 세트['data'] 를 이용하면 된다.
# 나머지 key가 가리키는 데이터 값의 추출도 동일하게 수행하면 된다.

keys = iris_data.keys()
print('붓꽃 데이터 세트의 키들:', keys)



```

```python
# 각 데이터 세트의 키가 가리키는 데이터 가리키는 값들을 나타내기

print('\n feature_names 의 type:',type(iris_data.feature_names))
print(' feature_names 의 shape:',len(iris_data.feature_names))
print(iris_data.feature_names)

print('\n target_names 의 type:',type(iris_data.target_names))
print(' feature_names 의 shape:',len(iris_data.target_names))
print(iris_data.target_names)

print('\n data 의 type:',type(iris_data.data))
print(' data 의 shape:',iris_data.data.shape)
print(iris_data['data'])

print('\n target 의 type:',type(iris_data.target))
print(' target 의 shape:',iris_data.target.shape)
print(iris_data.target)
```
