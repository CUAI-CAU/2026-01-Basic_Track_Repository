```python
# 넘파이, 판다스, 맷플롯립과 시본 불러오기

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

titanic_df = pd.read_csv('./titanic_train.csv')
titanic_df.head(3)
```

```python
# RangeIndex는 DataFrame 인덱스의 범위를 나타내므로 전체 로우 수를 알 수 있다.
# 칼럼수는 12개이고, 2개의 칼럼이 float64 타입, 5개의 칼럼이 object 타입이다.(string 타입)
# Age와 Cabin, Embarked 칼럼은 결측치를 가지는 것을 알 수 있다.
print('\n ### train 데이터 정보 ###  \n')
print(titanic_df.info())
```

```python
# 사이킷런 머신러닝 알고리즘은 Null 값을 허용하지 않으므로 Null 값을 처리해야 한다.
# Age칼럼은 평균 나이, 나머지 두 칼럼은 N으로 변경 
# inplace = True를 통해 원본 데이터 프레임의 내용으로 바로 수정

titanic_df['Age'].fillna(titanic_df['Age'].mean(),inplace=True)
titanic_df['Cabin'].fillna('N',inplace=True)
titanic_df['Embarked'].fillna('N',inplace=True)
print('데이터 세트 Null 값 갯수 ',titanic_df.isnull().sum().sum())
```

```python
# 남아있는 문자열 피처의 값 분류 살펴보기.
print(' Sex 값 분포 :\n',titanic_df['Sex'].value_counts())
print('\n Cabin 값 분포 :\n',titanic_df['Cabin'].value_counts())
print('\n Embarked 값 분포 :\n',titanic_df['Embarked'].value_counts())
```

```python
# 선실 등급을 나타내는 첫 번째 알파벳이 중요할 것이라 생각
# Cabin 칼럼의 경우 앞 문자만 추출

titanic_df['Cabin'] = titanic_df['Cabin'].str[:1]
print(titanic_df['Cabin'].head(3))
```

```python
# 성별이 생존 확률에 어떤 영향을 미쳤는지, 성별에 따른 생존자 수 비교.
# groupby를 통해 따로 분류하고 survived 열을 count한다.
# 0은 사망 1은 생조니다.
titanic_df.groupby(['Sex','Survived'])['Survived'].count()
```

```python
# 위의 결과를 통해 남자는 577명중 468명이 죽고 109명만 살아남아 약 18.8%가 생존
# 여자는 314명 중 233명으로 약 74.2%가 생존
# 시본 그래프로 확인
# sns.barplot은 디폴트 값으로 y축 데이터의 평균값을 구하도록 설계되어있음. (estimator 파라미터 디폴트 값이 numpy.mean 예를 들어 단순 합계를 보고싶으면 estimator = np.sum으로 바꾸면 된다.)
sns.barplot(x='Sex', y = 'Survived', data=titanic_df)
```

```python
# hue 파라미터를 추가해서 성별까지 함께 고려하여 그래프를 나타낼수도있다.
sns.barplot(x='Pclass', y='Survived', hue='Sex', data=titanic_df)
```

```python
# age에 따른 생존 확률을 알아보기
# Age는 값의 종류가 많기 때문에 범위별로 분류해서 카테고리 값 할당.
# 입력 age에 따라 구분값을 반환하는 함수 설정. DataFrame의 apply lambda식에 사용. 
def get_category(age):
    cat = ''
    if age <= -1: cat = 'Unknown'
    elif age <= 5: cat = 'Baby'
    elif age <= 12: cat = 'Child'
    elif age <= 18: cat = 'Teenager'
    elif age <= 25: cat = 'Student'
    elif age <= 35: cat = 'Young Adult'
    elif age <= 60: cat = 'Adult'
    else : cat = 'Elderly'
    
    return cat

# 막대그래프의 크기 figure를 더 크게 설정 
plt.figure(figsize=(10,6))

#X축의 값을 순차적으로 표시하기 위한 설정 
group_names = ['Unknown', 'Baby', 'Child', 'Teenager', 'Student', 'Young Adult', 'Adult', 'Elderly']

# lambda 식에 위에서 생성한 get_category( ) 함수를 반환값으로 지정. 
# get_category(X)는 입력값으로 'Age' 컬럼값을 받아서 해당하는 cat 반환
# Age_cat 이라는 새로운 피처 생성
titanic_df['Age_cat'] = titanic_df['Age'].apply(lambda x : get_category(x))
sns.barplot(x='Age_cat', y = 'Survived', hue='Sex', data=titanic_df, order=group_names)
# 시각화 용도로 만들었던 Age_cat을 원본 데이터프레임에서 삭제.
titanic_df.drop('Age_cat', axis=1, inplace=True)
```

```python
# 위의 분석 결과로 Sex, Age, PClass 등이 중요하게 생존을 좌우하는 피처임을 확인
# 남아있는 문자열 카테고리 피처를 숫자형 카테고리 피처로 변환.
# 사이킷런의 LabelEncoder 클래스를 이용해 레이블 인코딩을 적용.

from sklearn import preprocessing

def encode_features(dataDF):
    features = ['Cabin', 'Sex', 'Embarked']
    for feature in features:
        le = preprocessing.LabelEncoder()
        le = le.fit(dataDF[feature])
        dataDF[feature] = le.transform(dataDF[feature])
        
    return dataDF

titanic_df = encode_features(titanic_df)
titanic_df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

# Null 처리 함수
# Age열은 평균으로 채우고 Cabin, Embarked 열은 N으로 채우고, Fare 열은 0으로 채웠다
def fillna(df):
    df['Age'].fillna(df['Age'].mean(), inplace=True)
    df['Cabin'].fillna('N', inplace=True)
    df['Embarked'].fillna('N', inplace=True)
    df['Fare'].fillna(0, inplace=True)
    return df

# 머신러닝 알고리즘에 불필요한 피처 제거
def drop_features(df):
    df.drop(['PassengerId', 'Name', 'Ticket'], axis=1, inplace=True)
    return df

# 레이블 인코딩 수행.
# Cabin 열의 첫 문자만 뽑아낸후 레이블 인코딩
def format_features(df):
    df['Cabin'] = df['Cabin'].str[:1]
    features = ['Cabin', 'Sex', 'Embarked']
    for feature in features:
        le = LabelEncoder()
        le = le.fit(df[feature])
        df[feature] = le.transform(df[feature])
    return df

# 앞에서 설정한 데이터 전처리 함수 호출
def transform_features(df):
    df = fillna(df)
    df = drop_features(df)
    df = format_features(df)
    return df
```

```python
# 원본 데이터를 재로딩 하고, feature데이터 셋과 Label 데이터 셋 추출. 
titanic_df = pd.read_csv('./titanic_train.csv')
# survived 열만 떼오기 (정답)
y_titanic_df = titanic_df['Survived']
# survived 열 빼고 떼오기 (피처, 문제지)
X_titanic_df= titanic_df.drop('Survived',axis=1)

# 전에 정의했던 전처리 함수를 통해 결측치 인코딩등을 하는 것,
X_titanic_df = transform_features(X_titanic_df)
```

```python
# train_test_split()을 통해 테스트 세트 크기는 20%로 나눈다.
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test=train_test_split(X_titanic_df, y_titanic_df, \
                                                  test_size=0.2, random_state=11)
```

```python
# 결정 트리를 위해 DecisionTreeClassifier,
# 랜덤 포레스트를 위해 RandomForestClassifier
# 로지스틱 회귀를 위해 LogisticRegression 클래스 사용

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# 결정트리, Random Forest, 로지스틱 회귀를 위한 사이킷런 Classifier 클래스 생성
dt_clf = DecisionTreeClassifier(random_state=11)
rf_clf = RandomForestClassifier(random_state=11)
# solver = 'liblinear'는 로지스틱 회귀의 최적화 알고리즘을 liblinear로 설정
lr_clf = LogisticRegression(solver='liblinear')

# DecisionTreeClassifier 학습/예측/평가
dt_clf.fit(X_train , y_train)
dt_pred = dt_clf.predict(X_test)
print('DecisionTreeClassifier 정확도: {0:.4f}'.format(accuracy_score(y_test, dt_pred)))

# RandomForestClassifier 학습/예측/평가
rf_clf.fit(X_train , y_train)
rf_pred = rf_clf.predict(X_test)
print('RandomForestClassifier 정확도:{0:.4f}'.format(accuracy_score(y_test, rf_pred)))

# LogisticRegression 학습/예측/평가
lr_clf.fit(X_train , y_train)
lr_pred = lr_clf.predict(X_test)
print('LogisticRegression 정확도: {0:.4f}'.format(accuracy_score(y_test, lr_pred)))

# 3개의 알고리즘 중 로지스틱회귀가 타 알고리즘에 비해 높은 정확도를 나타냄.
```

```python
# 교차 검증으로 결정 트리 모델을 평가하기 (KFold)

from sklearn.model_selection import KFold

def exec_kfold(clf, folds=5):
    # 폴드 세트를 5개인 KFold객체를 생성, 폴드 수만큼 예측결과 저장을 위한  리스트 객체 생성.
    kfold = KFold(n_splits=folds)
    scores = []
    
    # KFold 교차 검증 수행. 
    for iter_count , (train_index, test_index) in enumerate(kfold.split(X_titanic_df)):
        # X_titanic_df 데이터에서 교차 검증별로 학습과 검증 데이터를 가리키는 index 생성
        X_train, X_test = X_titanic_df.values[train_index], X_titanic_df.values[test_index]
        y_train, y_test = y_titanic_df.values[train_index], y_titanic_df.values[test_index]
        
        # Classifier 학습, 예측, 정확도 계산 
        clf.fit(X_train, y_train) 
        predictions = clf.predict(X_test)
        accuracy = accuracy_score(y_test, predictions)
        scores.append(accuracy)
        print("교차 검증 {0} 정확도: {1:.4f}".format(iter_count, accuracy))     
    
    # 5개 fold에서의 평균 정확도 계산. 
    mean_score = np.mean(scores)
    print("평균 정확도: {0:.4f}".format(mean_score)) 
# exec_kfold 호출
exec_kfold(dt_clf , folds=5) 
```

```python
# cross_val_score()를 통한 교차검증

from sklearn.model_selection import cross_val_score

scores = cross_val_score(dt_clf, X_titanic_df, y_titanic_df, cv=5)
for iter_count, accuracy in enumerate(scores):
    print("교차 검증 {0} 정확도: {1:.4f}".format(iter_count, accuracy))

print('평균 정확도: {0:.4f}'.format(np.mean(scores)))
```

```python
# GridSearchCV를 이용해 결정트리모델의 최적 하이퍼 파라미터를 찾고 예측 성능 측정.

from sklearn.model_selection import GridSearchCV

# 나무의 깊이 (4가지 후보)
# 노드를 나누기 위한 최소 데이터 수
# 리프 노드(끝단)이 되기 위한 최소 데이터수
# cv = 5를 통해 5-폴드 교차 검즈 수행

parameters = {'max_depth':[2,3,5,10],
             'min_samples_split':[2,3,5], 'min_samples_leaf':[1,5,8]}

grid_dclf = GridSearchCV(dt_clf , param_grid=parameters , scoring='accuracy' , cv=5)
grid_dclf.fit(X_train , y_train)

print('GridSearchCV 최적 하이퍼 파라미터 :',grid_dclf.best_params_)
print('GridSearchCV 최고 정확도: {0:.4f}'.format(grid_dclf.best_score_))
best_dclf = grid_dclf.best_estimator_

# GridSearchCV의 최적 하이퍼 파라미터로 학습된 Estimator로 예측 및 평가 수행. 
dpredictions = best_dclf.predict(X_test)
accuracy = accuracy_score(y_test , dpredictions)
print('테스트 세트에서의 DecisionTreeClassifier 정확도 : {0:.4f}'.format(accuracy))
```
