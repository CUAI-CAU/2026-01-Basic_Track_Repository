```python
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split

iris = load_iris()

# data와 target 등의 feature로 구성되어 있다.
iris_data = iris['data']

iris_label = iris['target']

print('iris target names:', iris['target_names'])
print('iris target 값:', iris_label)

iris_df = pd.DataFrame(data=iris_data, columns=iris['feature_names'])
iris_df['label'] = iris['target']
#feature names와 target names 가 별도인가?
print(iris['target_names'])
print(iris['feature_names'])
iris_df.head(5)
```

```python
train_input, test_input, train_target, test_target = train_test_split(iris_data, iris_label, test_size=0.2, random_state=42)
```

```python
dt_clf = DecisionTreeClassifier(random_state=42)

dt_clf.fit(train_input, train_target)
```

```python
pred = dt_clf.predict(test_input)
pred
```

```python
from sklearn.metrics import accuracy_score
print('예측 정확도: {0:.4f}'.format(accuracy_score(test_target, pred)))
# 이 중괄호는 자리 표시자라는 기능을 한다. 
# 처음의 0은 인덱스로서, .format() 에 들어온 첫 번째 인자(즉, 여기서는 accuracy_score)를 의미한다.
# 구분자 :은 이제부터 이 안에 들어오는 데이터를 어떻게 보여줄지 형식을 지정한다.
# 정밀도 .4 는 소수점 아래 4자리까지 표시
# 타입 f는 부동 소수점(float) 형식으로 출력하라는 명이다.
```

```python
# 학습 데이터와 테스트 데이터를 분리 -> fit -> predict -> evaluation
# 03 사이킷런의 기반 프레임워크 익히기
```

```python
# 교차 검증
from sklearn.model_selection import KFold
import numpy as np

iris = load_iris()
features = iris['data']
label = iris['target']
dt_clf = DecisionTreeClassifier(random_state=42)

kfold = KFold(n_splits=5)
cv_accuracy = []
print(features)
print('붓꽃 데이터 세트 크기:', len(features))
# features.shape[0] 으로 행의 개수를 확인할 셈이라면, 그냥 len(features)를 써도 된다.
```

```python
n_iter = 0

for train_index, test_index in kfold.split(features):
    train_input, test_input = features[train_index], features[test_index]
    train_target, test_target = label[train_index], label[test_index]
    dt_clf.fit(train_input, train_target)
    pred = dt_clf.predict(test_input)
    n_iter += 1
    accuracy = np.round(accuracy_score(test_target, pred), 4)
    train_size = train_input.shape[0]
    test_size = test_input.shape[0]
    print('\n#{0} 교차 검증 정확도 :{1}, 학습 데이터 크기: {2}, 검증 데이터 크기: {3}'
          .format(n_iter, accuracy, train_size, test_size))
    print('#{0} 검증 세트 인덱스:{1}'.format(n_iter, test_index))
    cv_accuracy.append(accuracy)

print('\n## 평균 검증 정확도:', np.mean(cv_accuracy))
# for + format.. 코드를 작성하는 다양한 기법이 복합적으로 있다. 온전히 이해하기 보다는 여기서는 CV의 작동 원리만 파악하는 정도로.
```

```python
dt_clf = DecisionTreeClassifier(random_state=42)
skfold = StratifiedKFold(n_splits=3)
n_iter=0
cv_accuracy=[]

# StratifiedKFold의 split( ) 호출시 반드시 레이블 데이터 세트도 추가 입력 필요
for train_index, test_index in skfold.split(features, label):

# split( )으로 반환된 인덱스를 이용해 학습용, 검증용 테스트 데이터 추출
    train_input, test_input = features[train_index], features[test_index]
    train_target, test_target = label[train_index], label[test_index]

#학습 및 예측
    dt_clf.fit(train_input, train_target)
    pred = dt_clf.predict(test_input)

# 반복 시마다 정확도 측정
    n_iter += 1
    accuracy = np.round(accuracy_score(test_target, pred), 4)
    train_size = train_input.shape[0]
    test_size = test_input.shape[0]

print('\n#{0} 교차 검증 정확도 :{1}, 학습 데이터 크기: {2}, 검증 데이터 크기: {3}'
.format(n_iter, accuracy, train_size, test_size))
```

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

train_array = np.arange(0, 11).reshape(-1, 1)
test_array = np.arange(0, 6).reshape(-1, 1)
```

```python
scaler = MinMaxScaler()

scaler.fit(train_array)
# train_array의 특성을 가지고 scaler을 학습시킨다. 
# scaler의 프레임이 맞춰지는 순간이다.

train_scaled = scaler.transform(train_array)
# 학습시킨 scaler를 이제 활용하는 단계다.
# 새로운 객체 train_scaled 를 transform으로 바꿔서 새롭게 설정한다.

print('원본 train_array data:', np.round(train_array.reshape(-1), 2))
print('Scale된 train_array data:', np.round(train_scaled.reshape(-1), 2))

# round의 2번째 매개변수는 삭제시킬 자릿수를 설정한다. 그러므로 2번째 자리에서
# 삭제시키고 1번째까지 출력된다.
# fit 하고 transform 까지는 2차원만 가능하므로, 결과를 출력할 때는 1차원으로
```

```python
scaler.fit(test_array)

test_scaled = scaler.transform(test_array)

print('원본 test_array data:', np.round(test_array.reshape(-1), 2))
print('Scale된 test_array data:', np.round(test_scaled.reshape(-1), 2))

# test_array에 scale 변환할 때는 fit() 을 절대 쓰지 않고, transform() 만으로 
# 변환. 그러므로 fit_transform()은 학습 데이터에만 사용하고, 테스트 데이터에는
# 사용하지 않는다.

```

```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV

# 1. 가상의 데이터 (이미 학습용과 테스트용으로 분리되었다고 가정)
# 데이터 형태: [연소득, 연체횟수]
train_input = np.array([[5000, 1], [3000, 4], [8000, 0], [4500, 2]])
# 1: 상환, 0: 미상환
train_target = np.array([1, 0, 1, 1]) 

test_input = np.array([[4000, 3], [6000, 1]])
test_target = np.array([0, 1])

# 2. scaler 객체 설정
scaler = StandardScaler()

# 빈칸 (A): train 데이터를 기준으로 평균과 표준편차를 학습(fit)시킵니다.
scaler.fit(train_input)

# 학습된 기준으로 train 데이터를 변환합니다.
train_scaled = scaler.transform(train_input)
print(train_scaled)

# 빈칸 (B): test 데이터를 변환합니다.
test_scaled = scaler.transform(test_input)

# 3. 모델 및 하이퍼파라미터 설정
dt = DecisionTreeClassifier(random_state=42)
param_grid = {'max_depth': [3, 5, 7], 'min_samples_split': [2, 3, 4]}

# 4. GridSearchCV 적용
# 빈칸 (C): 튜닝할 모델, 파라미터 조합, 그리고 교차 검증 3회를 설정하여 객체를 만듭니다.
grid_dt = GridSearchCV(estimator=dt, cv=3, n_jobs=-1, param_grid=param_grid)

# 빈칸 (D): 준비된 GridSearchCV 객체에 스케일링된 훈련 데이터와 타겟 데이터를 넣어 학습시킵니다.
grid_dt.fit(train_scaled, train_target)

print("최적의 파라미터:", grid_dt.best_params_)
score_df = pd.DataFrame(grid_dt.cv_results_)
score_df.sort_values('rank_test_score')
```
