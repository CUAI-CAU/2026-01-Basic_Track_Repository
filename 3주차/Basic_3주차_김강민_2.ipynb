```python
# load_iris() 함수를 불러온다.
from sklearn.datasets import load_iris

# 붓꽃 데이터 세트를 로드하여 iris_data 변수에 저장한다.
iris_data = load_iris()

# 불러온 데이터의 자료형을 출력한다.
print(type(iris_data))
```

```python
# 붓꽃 데이터 세트가 가지고 있는 key 값을 확인한다.
keys = iris_data.keys()

# key 목록을 출력한다.
print('붓꽃 데이터 세트의 키들:', keys)
```

```python
# feature_names의 자료형과 개수, 실제 내용을 출력한다.
# feature_names는 각 열(column)이 어떤 특성을 의미하는지 보여준다.
print('\n feature_names 의 type:', type(iris_data.feature_names))
print(' feature_names 의 shape:', len(iris_data.feature_names))
print(iris_data.feature_names)

# target_names의 자료형과 개수, 실제 내용을 출력한다.
# target_names는 붓꽃 품종의 이름을 나타낸다.
print('\n target_names 의 type:', type(iris_data.target_names))
print(' target_names 의 shape:', len(iris_data.target_names))
print(iris_data.target_names)

# data의 자료형과 shape, 실제 데이터를 출력한다.
# data는 붓꽃의 각 샘플에 대한 특성값을 담고 있다.
print('\n data 의 type:', type(iris_data.data))
print(' data 의 shape:', iris_data.data.shape)
print(iris_data['data'])

# target의 자료형과 shape, 실제 값을 출력한다.
# target은 각 데이터가 어떤 품종인지 숫자로 나타낸 정답(label)이다.
print('\n target 의 type:', type(iris_data.target))
print(' target 의 shape:', iris_data.target.shape)
print(iris_data.target)
```
