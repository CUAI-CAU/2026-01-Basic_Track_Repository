### DataFrame과 리스트, 딕셔너리, 넘파이 ndarray 상호 변환

* 넘파이 ndarray, 리스트, 딕셔너리를 DataFrame으로 변환하기

```python
# 수치 배열 생성을 위해 넘파이를 불러옵니다.
import numpy as np

# 1차원 데이터를 DataFrame으로 바꿀 때 사용할 단일 컬럼명을 준비합니다.
col_name1 = ['col1']
list1 = [1, 2, 3]

# 리스트를 ndarray로 변환해 배열 형태도 함께 확인합니다.
array1 = np.array(list1)
print('array1 shape:', array1.shape)

# 1차원 리스트를 DataFrame으로 변환합니다.
df_list1 = pd.DataFrame(list1, columns=col_name1)
print('1차원 리스트로 만든 DataFrame:\n', df_list1)

# 1차원 ndarray도 동일하게 DataFrame으로 변환할 수 있습니다.
df_array1 = pd.DataFrame(array1, columns=col_name1)
print('1차원 ndarray로 만든 DataFrame:\n', df_array1)
```

```python
# 2차원 데이터는 열이 3개이므로 컬럼명도 3개를 지정합니다.
col_name2 = ['col1', 'col2', 'col3']

# 2행 3열 구조의 리스트를 생성합니다.
list2 = [[1, 2, 3],
         [11, 12, 13]]

# 동일한 데이터를 ndarray로도 만들어 둡니다.
array2 = np.array(list2)
print('array2 shape:', array2.shape)

# 2차원 리스트를 DataFrame으로 변환합니다.
df_list2 = pd.DataFrame(list2, columns=col_name2)
print('2차원 리스트로 만든 DataFrame:\n', df_list2)

# 2차원 ndarray도 같은 방식으로 DataFrame으로 변환합니다.
df_array2 = pd.DataFrame(array2, columns=col_name2)
print('2차원 ndarray로 만든 DataFrame:\n', df_array2)
```

```python
# 딕셔너리의 key는 컬럼명, value는 컬럼 데이터로 사용됩니다.
dict = {'col1': [1, 11], 'col2': [2, 22], 'col3': [3, 33]}

# 딕셔너리로부터 DataFrame을 생성합니다.
df_dict = pd.DataFrame(dict)
print('딕셔너리로 만든 DataFrame:\n', df_dict)
```

* DataFrame을 넘파이 ndarray, 리스트, 딕셔너리로 변환하기

```python
# DataFrame의 values 속성을 사용하면 내부 데이터를 ndarray로 얻을 수 있습니다.
array3 = df_dict.values
print('df_dict.values 타입:', type(array3), 'df_dict.values shape:', array3.shape)

# 변환된 ndarray 내용을 출력합니다.
print(array3)

```

```python
# ndarray를 다시 파이썬 리스트 형태로 바꿉니다.
list3 = df_dict.values.tolist()
print('df_dict.values.tolist() 타입:', type(list3))
print(list3)

# DataFrame을 딕셔너리로 변환할 때 orient='list' 형식을 사용합니다.
# 각 key는 컬럼명이고 value는 해당 컬럼의 리스트가 됩니다.
dict3 = df_dict.to_dict('list')
print('\n df_dict.to_dict() 타입:', type(dict3))
print(dict3)
```

### DataFrame의 컬럼 데이터 셋 Access

```python
# 새 컬럼 Age_0을 추가하고 모든 행 값을 0으로 채웁니다.
titanic_df['Age_0'] = 0

# 새 컬럼이 잘 추가되었는지 상위 3개 행으로 확인합니다.
titanic_df.head(3)
```

```python
# Age 값을 10배 한 파생 컬럼을 생성합니다.
titanic_df['Age_by_10'] = titanic_df['Age'] * 10

# 형제자매 수(SibSp), 부모자녀 수(Parch), 본인 1명을 합쳐 가족 수를 계산합니다.
titanic_df['Family_No'] = titanic_df['SibSp'] + titanic_df['Parch'] + 1

# 계산된 새 컬럼들을 확인합니다.
titanic_df.head(3)
```

```python
# 기존 파생 컬럼 Age_by_10에 100을 더해 값을 일괄 수정합니다.
titanic_df['Age_by_10'] = titanic_df['Age_by_10'] + 100

# 수정 결과를 상위 3개 행으로 확인합니다.
titanic_df.head(3)
```

* 불린 인덱싱

```python
# 불린 인덱싱 예제를 위해 원본 데이터를 다시 읽습니다.
titanic_df = pd.read_csv('titanic_train.csv')

# 나이가 60세 초과인 승객만 조건식으로 필터링합니다.
titanic_boolean = titanic_df[titanic_df['Age'] > 60]

# 불린 인덱싱 결과도 DataFrame 객체입니다.
print(type(titanic_boolean))
titanic_boolean
```

```python
# 조건 필터링 후 필요한 컬럼(Name, Age)만 다시 선택합니다.
titanic_df[titanic_df['Age'] > 60][['Name', 'Age']].head(3)
```

```python
# 같은 작업을 loc로 작성하면 행 조건과 열 선택을 한 번에 표현할 수 있습니다.
titanic_df.loc[titanic_df['Age'] > 60, ['Name', 'Age']].head(3)
```

```python
# 여러 조건을 &, | 연산자로 결합해 복합 조건 필터링을 수행합니다.
# 여기서는 60세 초과이면서 1등실이면서 여성인 승객만 선택합니다.
titanic_df[(titanic_df['Age'] > 60) & (titanic_df['Pclass'] == 1) & (titanic_df['Sex'] == 'female')]
```

```python
# 복합 조건을 개별 변수로 분리하면 가독성과 재사용성이 좋아집니다.
cond1 = titanic_df['Age'] > 60
cond2 = titanic_df['Pclass'] == 1
cond3 = titanic_df['Sex'] == 'female'

# 분리한 조건을 다시 결합해 같은 필터링 결과를 얻습니다.
titanic_df[cond1 & cond2 & cond3]
```

### apply lambda 식으로 데이터 가공

```python
# 일반 함수로 제곱 계산 로직을 정의합니다.
def get_square(a):
    return a ** 2

# 함수 호출 결과를 확인합니다.
print('3의 제곱은:', get_square(3))
```

```python
# 같은 제곱 연산을 lambda 식으로 간단히 표현합니다.
lambda_square = lambda x: x ** 2
print('3의 제곱은:', lambda_square(3))
```

```python
# map과 lambda를 함께 사용해 리스트 각 원소의 제곱을 계산합니다.
a = [1, 2, 3]
squares = map(lambda x: x ** 2, a)

# map 객체를 리스트로 변환해 결과를 확인합니다.
list(squares)
```

```python
# 각 승객 이름 문자열의 길이를 계산해 Name_len 컬럼으로 저장합니다.
titanic_df['Name_len'] = titanic_df['Name'].apply(lambda x: len(x))

# 원본 이름과 계산된 길이를 함께 확인합니다.
titanic_df[['Name', 'Name_len']].head(3)
```

```python
# Age 값을 기준으로 15세 이하면 Child, 그보다 크면 Adult로 분류합니다.
titanic_df['Child_Adult'] = titanic_df['Age'].apply(lambda x: 'Child' if x <= 15 else 'Adult')

# 나이와 분류 결과를 함께 확인합니다.
titanic_df[['Age', 'Child_Adult']].head(8)
```

```python
# 중첩 삼항 연산자를 사용해 나이를 Child, Adult, Elderly 세 범주로 나눕니다.
titanic_df['Age_cat'] = titanic_df['Age'].apply(
    lambda x: 'Child' if x <= 15 else ('Adult' if x <= 60 else 'Elderly')
)

# 각 범주의 빈도를 집계해 분류 결과를 확인합니다.
titanic_df['Age_cat'].value_counts()
```

```python
# 나이에 따라 세분화된 범주를 반환하는 사용자 정의 함수를 만듭니다.
def get_category(age):
    cat = ''
    if age <= 5:
        cat = 'Baby'
    elif age <= 12:
        cat = 'Child'
    elif age <= 18:
        cat = 'Teenager'
    elif age <= 25:
        cat = 'Student'
    elif age <= 35:
        cat = 'Young Adult'
    elif age <= 60:
        cat = 'Adult'
    else:
        cat = 'Elderly'

    return cat

# apply()로 Age 컬럼의 각 값에 사용자 정의 함수를 적용합니다.
titanic_df['Age_cat'] = titanic_df['Age'].apply(lambda x: get_category(x))

# 나이와 분류 결과를 함께 확인합니다.
titanic_df[['Age', 'Age_cat']].head()
```
