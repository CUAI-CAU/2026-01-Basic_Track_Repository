# 데이터 추출 방법: 넘파이 vs 판다스

*데이터의 저장 방식*
- 넘파이: ndarry(n차원)
> 수치 연산에 강함 : 위치
- 판다스: Series(1차원), **DataFrame(2차원 이상)**
> 데이터 분석에 강함 : 위치 + 명칭
>> 수치 연산에는 약함(불편?)

**23pg. (Numpy_ndarray) 넘파이의 ndarray의 데이터 세트 선택하기 - 인덱싱**
1. 특정한 데티어만 추출: 원하는 위치의 인덱스 값을 지정하면 해당 위치의 데이터가 반환
2. 슬라이싱: ':' 리스트
3. 팬시 인덱싱: 인덱스 집합
4. 불린 인덱싱: 특정 조건에 해당 여부 > True/False
- 1번의 경우 특정 위치, 2번은 리스트, 3번은 인덱스의 집합을 불러오기

**62pg. (Pandas_DataFrame) 데이터 셀렉션 및 필터링**

> ['칼럼명']: 1개 열 / [['칼럼명', '칼럼명']]: 2개 이상의 열

# 6.1. DataFrame의 [ ]연산자 [정리]
1. 칼럼명: 열(칼럼_column)으로 찾음 (일반적인 case)

2. 숫자(일반): 일반적으로 오류 발생
3. 숫자(예외1): 슬라이싱':' > 행(row)로 찾음
4. 숫자(예외2): 불린 인덱싱

#Tip. 일반적으로 사용하지 않는 것이 좋다!

# 7.3. Groupby (numpy에는 없는 놈)
> 새로운 기준으로 재편성 : 그 기준이 곧 '칼럼'

1. titanic_df.groupby(by='Pclass')
> Pclass를 기준으로 재편성

2. titanic_df.groupby('Pclass').count()
> 재편성 기준으로 각각의 칼럼 총계

3. titanic_df.groupby('Pclass')[['PassengerId','Survived']].count()
> 재편성한 것들 중에서 특정 칼럼만 추출

4. groupby('Pclass')['Age'].agg([max,min])
> 추출한 칼럼들의 max, min

```python
#7.3.1 groupby()적용 - Numpy에 없음
#groupby란? 새로운 그룹으로 묶어주는 것
#: 데이터를 새로운 기준으로 "재편성"
titanic_groupby = titanic_df.groupby(by='Pclass')
print(type(titanic_groupby))
#Pclass(1,2,3)로 그룹지어준 것 (Pclass = 1인 그룹...)
```

```python
#7.3.2 특정 칼럼에 대해서 그룹핑을 요구하는 경우에는 대상 칼럼을 기준으로 나머지 변수들에 aggregation 함수를 적용함
titanic_groupby = titanic_df.groupby('Pclass').count()
titanic_groupby
```

```python
#7.3.3 앞서 특정칼럼을 기준으로 나열한 것에서 몇가지만 빼오기
titanic_groupby = titanic_df.groupby('Pclass')[['PassengerId','Survived']].count()
titanic_groupby
```

```python
#7.3.4 앞서 뺴온 것에서 max, min 구하기
titanic_df.groupby('Pclass')['Age'].agg([max,min])
```

```python
#7.3.5 DataFrame groupby() 예제
agg_format={'Age':'max', 'SibSp':'sum', 'Fare':'mean'}
titanic_df.groupby('Pclass').agg(agg_format)
```

# 9 apply lambda 식으로 데이터 가공

# 9. 데이터 가공하기
> pandas의 무기 : 사실 데이터 분석을 하기 위해서는 데이터 가공이 필수적

***return(python 자체 언어) <-> lambda(pandas)***
> lambda x : x**2
>> 입력값 : 계산 과정

```python
#9.1.1 return을 사용하는 경우
def get_square(a):
    return a**2

print('3의 제곱은:', get_square(3))
```

```python
#9.1.2 lambda를 쓰는 경우~!
lambda_square = lambda x : x**2
print('3의 제곱은:', lambda_square(3))
# ':'으로 입력인자와 계산식을 분리시켜줌
```

```python
#9.1.3 map가 병합하는 경우
a=[1,2,3]
squares = map(lambda x : x**2, a)
list(squares)
#map: 하나씩 자동으로 넣어주기
```

```python
#9.2.1 DataFrame과 lambda의 병합 - 2차원에 적용!
titanic_df['Name_len']=titanic_df['Name'].apply(lambda x : len(x))
titanic_df[['Name', 'Name_len']].head(3)
```

```python
#9.2.2 lambda에 if / else절을 사용
titanic_df['Child_Adult'] = titanic_df['Age'].apply(lambda x : 'Child' if x <= 15 else 'Adult')
titanic_df[['Age', 'Child_Adult']].head(8)
```

```python
#9.2.3 lambda에서는 else if는 지원X >> 다음과 같이 귀찮게 해야함ㅋㅋ
titanic_df['Age_cat'] = titanic_df['Age'].apply(lambda x : 'Child' if x<=15 else ('Adult' if x <= 60
                                                                                  else 'Elderly'))
titanic_df['Age_cat'].value_counts()
```
