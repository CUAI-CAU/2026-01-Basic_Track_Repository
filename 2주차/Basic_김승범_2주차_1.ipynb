### 결손 데이터 처리하기

```python
titanic_df.isna().head(3) #isna로 널값인지 아닌지 반환해줌.
```

```python
titanic_df.isna().sum()
```

```python
#fillna로 결손값 대체하기
titanic_df['Cabin']=titanic_df['Cabin'].fillna('C000')
titanic_df.head(3)
```

```python
#age값의 NaN을 평균 나이로, Embarked 칼럽의 널 값을 S로 대체하여 결손데이터 처리하는 예제
titanic_df['Age']=titanic_df['Age'].fillna(titanic_df['Age'].mean())
titanic_df['Embarked']=titanic_df['Embarked'].fillna('S')
titanic_df.isna().sum()
```

```python
### apply lambda 식
lambda_square=lambda x:x**2 # :보다 왼쪽에 있는게 입력인자, 오른쪽에 있는게 출력인자

print('3의 제곱 근은:', lambda_square(3))
```

```python
#여러개의 값을 인자로 받아야 할 때
a=[1,2,3]
squares=map(lambda x:x**2,a)
list(squares)
```

```python
titanic_df['Name_len']=titanic_df['Name'].apply(lambda x:len(x))
titanic_df[['Name','Name_len']].head(3)
```

```python
titanic_df['Child_Adult']=titanic_df['Age'].apply(lambda x: 'Child' if x<=15 else 'Adult')
#:오른쪽에 반환 값이 나와야 해서 if문이 먼저 나오는게 아니라 반환 값이 먼저 나와야함
#그리고 if/else 밖에 안됨. else if쓰려면 else문 안에다가 또 조건문 넣어야만 함.
#그냥 파이썬 리스트는 apply안써도 되지만 여기처럼 

titanic_df[['Age', 'Child_Adult']].head(8)
                                                  
```
