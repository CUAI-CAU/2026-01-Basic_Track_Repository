```python
import matplotlib.pyplot as plt
```

```python
# x축 입력값 [1.2.3], y축 입력값 [2,4,6]으로 선그래프 생성
plt.plot([1,2,3],[2,4,6])

# 선그래프 타이틀을 hello plot으로 설정
plt.title('hello plot')

# 선그래프 출력
plt.show()
```

```python
# plt.figure()은 주로 figure의 크기를 조절하는 데 사용됨
plt.figure(figsize=(10,4))

plt.plot([1,2,3],[2,4,6])
plt.title('hello plot')
plt.show()
```

```python
plt.figure(figsize=(10,4), facecolor='yellow') # 그림판 배경색 노랑으로
plt.plot([1,2,3],[2,4,6])
plt.title('hello plot')
plt.show()
```

```python
# pyplot에서 설정된 figure와 axes 객체를 함께 가져오기
fig, ax = plt.subplots()

# Axes 객체의 plot()을 이용하여 선그래프를 그림
ax.plot([1,2,3],[2,4,6])

# Axes 객체의 set_title()을 이용하여 선그래프 타이틀을 hello plot으로 설정
ax.set_title('hello plot')
plt.show()
```

```python
# axes 객체의 plot()과 set_title()을 이용해서 앞의 pyplot 과 title()을 이용해 그린 선그래프와 동일한 결과를 얻음
```

```python
## subplot으로 여러개의 plot 생성하기
# 개별 subplot은 하나의 axes를 가지게 됨
# plt.subplots()는 여러 개의 그래프르 그릴 수 있게 하는 1개의 figure와 여러 개의 axes를 생성하고 반환

# nrows 는 전체 subplot들의 배치를 2차원 행렬형태로 표현할 때 행의 개수, ncols는 열의 개수

# 두개의 subplot,즉 두개의 axes를 생성하되 열 방향으로 두개를 배치함
# 전체 크기가 가로 6, 세로 3인 figure 생성
fig, (ax1,ax2) = plt.subplots(nrows=1, ncols=2, figsize=(6,3))
```

```python
import numpy as np
import pandas as pd

x_value = [1,2,3,4]
y_value = [2,4,6,8]

# 입력값으로 파이썬 리스트, numpy array 가능, x 좌표값과 y좌표값은 모두 같은 크기를 가져야함
# 선 색깔을 녹색으로, 선의 형태는 대시형, 두께 2, 좌표별 동그라미 형태 마커의 크기를 12로 설정
plt.plot(x_value,y_value,color='green',marker='o',linestyle='dashed',linewidth=2,markersize=12)

# x축과 y축의 이름을 텍스트로 할당하여 설정
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.show()
```

```python
# x 좌표 값을 0~99까지 100개로 설정
x_value = np.arange(0,100)
# y 좌표 값은 개별 x 좌표값을 2배로 해서 100개로 설정
y_value = 2*x_value

plt.plot(x_value, y_value, color='green')
plt.xlabel('x axis')
plt.ylabel('y axis')

# x축 눈금값을 90도로 회전
plt.xticks(rotation=90)
plt.show()
```

```python
x_value = np.arange(0,100)
y_value = 2*x_value

plt.plot(x_value, y_value, color='green')
plt.xlabel('x axis')
plt.ylabel('y axis')

# x축 눈금값을 np.arange(0,100,5)로 하여 0부터 99까지 5스텝으로 20개를 표시하고
# 90도 회전
plt.xticks(ticks=np.arange(0,100,5),rotation=90)

plt.title('helloplot')
plt.show()
```

```python
# 범례(legend)는 여러개의 그래프/차트를 한 axes 내에서 표시할 때 편리하게 이용됨

# x좌표값은 0~99까지 100개, 첫번째 y좌표값은 x좌표값의 두배, 두번째 좌표값은 x좌표값의 4배
x_value1 = np.arange(0,100)
y_value1 = 2*x_value1
y_value2 = 4*x_value1

# x좌표값과 첫번째 y좌표값으로 녹색의 선그래프를 그림, label='green line'으로 설정, 두번째는 빨간색으로
plt.plot(x_value1,y_value1,color='green',label='greenline')
plt.plot(x_value1,y_value2,color='red',label='redline')

plt.xlabel('x axis')
plt.ylabel('y axis')
# 개별 plot들의 label로 설정된 문자열을 범례로 표시
plt.legend()

plt.show()
```

```python
x_value1 = np.arange(1,10)
y_value1 = 2*x_value1

# 마커를 포함한 빨간색 대시 선그래프
plt.plot(x_value1,y_value1,color='red',marker='o',linestyle='dashed',label='redline')
# x값에 따른 y값을 나타내는 초록색 막대 그래프를 그리고 label을 barplot으로 설정
plt.bar(x_value1,y_value1,color='green',label='barplot')

plt.xlabel('x axis')
plt.ylabel('y axis')
# 개별 plot들의 label로 설정된 문자열을 범례로 표시
plt.legend()

plt.show()
```

```python
import numpy as np

x_value1 = np.arange(1,10)
x_value2 = np.arange(1,20)
y_value1 = 2*x_value1
y_value2 = 2*x_value2

# figsize는 (6,3)인 figure와 2개의 axes를 가지는 subplots를 반환
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(6,3))

# 개별 axes 객체들을 각각 이용하여 선그래프와 막대그래프를 그림
ax[0].plot(x_value1,y_value1, color='red',marker='o',linestyle='dashed',label = 'redline')
ax[1].bar(x_value2,y_value2, color='green',label='barplot')

# 개별 axes 객체에 각각 x축명을 설정
ax[0].set_xlabel('ax[0] x axis')
ax[1].set_xlabel('ax[1] x axis')

# 개별 axes 객체 각각에 범례를 설정
ax[0].legend()
ax[1].legend()
# 개별 axes 객체 각각에 타이틀을 설정
ax[0].set_title('helloline')
ax[1].set_title('hellobar')

plt.show()
```

```python
import numpy as np

x_value1 = np.arange(1,10)
x_value2 = np.arange(1,20)
y_value1 = 2*x_value1
y_value2 = 2*x_value2

# figsize는 (6,6)인 figure와 2x2개의 axes를 가지는 subplots를 반환
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(6,6))

# 개별 axes 객체들을 각각 이용하여 선그래프와 막대그래프를 그림
ax[0][0].plot(x_value1,y_value1, color='red',marker='o',linestyle='dashed',label = 'redline')
ax[0][1].bar(x_value2,y_value2, color='green',label='barplot')
ax[1][0].plot(x_value1,y_value1, color='green',marker='o',linestyle='dashed',label = 'greenline')
ax[1][1].bar(x_value2,y_value2, color='red',label='redbar')

# 개별 axes 객체에 각각 x축명을 설정
ax[0][0].set_xlabel('ax[0][0] x axis')
ax[0][1].set_xlabel('ax[0][1] x axis')
ax[1][0].set_xlabel('ax[1][0] x axis')
ax[1][1].set_xlabel('ax[1][1] x axis')

# 개별 axes 객체 각각에 범례를 설정
ax[0][0].legend()
ax[0][1].legend()
ax[1][0].legend()
ax[1][1].legend()

plt.show()
```
