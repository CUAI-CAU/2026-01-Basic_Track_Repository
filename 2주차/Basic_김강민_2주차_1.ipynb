### DataFrame과 리스트, 딕셔너리, 넘파이 ndarray 상호 변환

* 넘파이 ndarray, 리스트, 딕셔너리를 DataFrame으로 변환하기

```python
import numpy as np

# 컬럼 이름 지정
col_name1=['col1']

# 리스트 생성
list1 = [1, 2, 3]

# numpy 배열 생성
array1 = np.array(list1)

print('array1 shape:', array1.shape )

# 리스트로 DataFrame 생성
df_list1 = pd.DataFrame(list1, columns=col_name1)
print('1차원 리스트로 만든 DataFrame:\n', df_list1)

# ndarray로 DataFrame 생성
df_array1 = pd.DataFrame(array1, columns=col_name1)
print('1차원 ndarray로 만든 DataFrame:\n', df_array1)
```

```python
# 컬럼 이름 3개 생성
col_name2=['col1', 'col2', 'col3']

# 2차원 리스트 생성
list2 = [[1, 2, 3],
         [11, 12, 13]]

# ndarray 생성
array2 = np.array(list2)

print('array2 shape:', array2.shape )

# 리스트 → DataFrame
df_list2 = pd.DataFrame(list2, columns=col_name2)
print('2차원 리스트로 만든 DataFrame:\n', df_list2)

# ndarray → DataFrame
df_array2 = pd.DataFrame(array2, columns=col_name2)
print('2차원 ndarray로 만든 DataFrame:\n', df_array2)
```

```python
# key는 컬럼명, value는 데이터
dict = {'col1':[1, 11], 'col2':[2, 22], 'col3':[3, 33]}

# 딕셔너리 → DataFrame 변환
df_dict = pd.DataFrame(dict)

print('딕셔너리로 만든 DataFrame:\n', df_dict)
```

* DataFrame을 넘파이 ndarray, 리스트, 딕셔너리로 변환하기

```python
# DataFrame → ndarray 변환
array3 = df_dict.values

print('df_dict.values 타입:', type(array3),
      'df_dict.values shape:', array3.shape)

print(array3)
```

```python
# DataFrame → 리스트 변환
list3 = df_dict.values.tolist()

print('df_dict.values.tolist() 타입:', type(list3))
print(list3)

# DataFrame → 딕셔너리 변환
dict3 = df_dict.to_dict('list')

print('\n df_dict.to_dict() 타입:', type(dict3))
print(dict3)
```

### apply lambda 식으로 데이터 가공

```python
# 제곱값을 반환하는 일반 함수 정의
def get_square(a):
    return a**2

# 일반 함수 실행
print('3의 제곱은:', get_square(3))
```

```python
# lambda 함수로 동일한 기능 구현
lambda_square = lambda x : x ** 2

# lambda 함수 실행
print('3의 제곱은:', lambda_square(3))
```

```python
# 리스트 생성
a = [1, 2, 3]

# map과 lambda를 이용해 각 원소 제곱
squares = map(lambda x : x**2, a)

# 결과를 리스트로 변환
list(squares)
```

```python
# Name 컬럼 문자열 길이 계산
titanic_df['Name_len'] = titanic_df['Name'].apply(lambda x : len(x))

# Name과 Name_len 컬럼 확인
titanic_df[['Name', 'Name_len']].head(3)
```

```python
# Age가 15 이하이면 Child, 아니면 Adult로 구분
titanic_df['Child_Adult'] = titanic_df['Age'].apply(
    lambda x : 'Child' if x <= 15 else 'Adult'
)

# 결과 확인
titanic_df[['Age', 'Child_Adult']].head(8)
```

```python
# Age 값에 따라 Child, Adult, Elderly로 구분
titanic_df['Age_cat'] = titanic_df['Age'].apply(
    lambda x : 'Child' if x <= 15 else ('Adult' if x <= 60 else 'Elderly')
)

# Age_cat 값 개수 확인
titanic_df['Age_cat'].value_counts()
```

```python
# 나이에 따라 세부 카테고리를 반환하는 함수 정의
def get_category(age):
    cat = ''

    if age <= 5:
        cat = 'Baby'
    elif age <= 12:
        cat = 'Child'
    elif age <= 18:
        cat = 'Teenager'
    elif age <= 25:
        cat = 'Student'
    elif age <= 35:
        cat = 'Young Adult'
    elif age <= 60:
        cat = 'Adult'
    else:
        cat = 'Elderly'

    return cat

# Age 컬럼에 함수 적용하여 Age_cat 컬럼 생성
titanic_df['Age_cat'] = titanic_df['Age'].apply(lambda x : get_category(x))

# 결과 확인
titanic_df[['Age', 'Age_cat']].head()
```
