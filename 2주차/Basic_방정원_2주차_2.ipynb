```python
# 2.1 맷플롯립의 pyplot 모듈의 이해 #matlab과 유사, 것보다는 조금 불편
#plot() : 선 그려주기
#title() : 제목
#show() : 화면에 출력

##pyploy모듈을 (관용적으로) plt라는 이름으로 import.
import matplotlib.pyplot as plt

##x축 입력값 [1,2,3], y축 입력값 [2,4,6]으로 선그래프 생성.
plt.plot([1,2,3],[2,4,6])

##선그래프 타이틀을 Hellp plot으로 설정.
plt.title("Hello plot")

##선 그래프를 출력
plt.show()
```

```python
#2.6.2 눈금을 회전 _ xticks(rotation=90)
##x 좌표 값을 0~99까지 100개로 설정.
x_value=np.arange(0,100)
##y 좌표 값은 개별 x 좌값을 2배로 해서 100개로 설정.
y_value=2*x_value

plt.plot(x_value, y_value, color='green')
plt.xlabel('x axis')
plt.ylabel('y axis')

##x축 눈금값을 090도로 회전
plt.xticks(rotation=90)
plt.show()
```

```python
#2.6.3 눈금의 개수를 수정  _ ticks=np.arange(0,100,5)
x_value=np.arange(0,100)
y_value=2*x_value

plt.plot(x_value, y_value, color='green')
plt.xlabel('x axis')
plt.ylabel('y axis')

##x축 눈금값을 090도로 회전
plt.xticks(ticks=np.arange(0,100,5), rotation=90)

plt.title('Hello plot')
plt.show()
```

```python
#2.6.4 한 그래프에 여러 값 _ plt.legend()
#X좌표값은 0~99까지 100개, 첫번째 Y좌표값은 X좌표값의 두 배, 두번째 좌표값은 X좌표값의 4배
x_value_01 =np.arange(0,100)
y_value_01 =2*x_value_01
y_value_02 =4*x_value_01

##X좌표값과 첫번째 Y좌표값으로 녹색의 선그래프를 스림. label을 'green line'으로 설정
plt.plot(x_value_01,y_value_01 ,color='green', label='green line')
##X좌표값과 두번째 Y좌표값으로 빨간색의 선그래프를 스림. label을 'red line'으로 설정
plt.plot(x_value_01,y_value_02 ,color='red', label='red line')

plt.xlabel('x axis')
plt.ylabel('y axis')

##개별 plot들의 label로 설정된 문자열을 범례로 표시 
plt.legend()
plt.title('Hello plot')
plt.show
```

```python
#2.6.5 막대그래프를 추가(1) _ plt.bar() version
x_value_01 = np.arange(1,10)
y_value_01 = 2*x_value_01

#마커를 포함한 빨간선 대시 선 그래프를 그리고, label은 'red line'으로 설정
plt.plot(x_value_01,y_value_01 ,color='green', label='green line')
plt.bar(x_value_01,y_value_01 ,color='red', label = 'red ploe')

#두 개의 그래프에서 label 설정된 문자열 값을 범례로 표시
plt.legend()

plt.title('Hello Plot')

plt.show
```

```python
#2.6.5 막대그래프를 추가(1) _ ax.bar() version
figure = plt.figure()
#Axes 객체를 추출하여 ax 변수에 할당.
ax=plt.axes()

#plt.plot()은 ax.plot()으로 plt.bar()는 ax.bar()로 변경.
ax.plot(x_value_01,y_value_01 ,color='green', label='green line')
ax.bar(x_value_01,y_value_01 ,color='red', label = 'red ploe')

#plt.xlabel()은 ax.set_xlabel()로, plt.ylabel()은 ax.set_ylabel()로 변경
ax.set_xlabel('x axis')
ax.set_ylabel('y axis')

ax.legend()

ax.set_title('Hello Plot')

plt.show
```

```python
#3.9 subplots를 이용하여 시본의 다양한 그래프를 시각화
## 맷플롯립 vs 시본(seaborn)
# 맷플롯립: "도화지(ax)야, 네가 직접 그려라"
# 시본: "시본아, 이 도화지(ax) 위에 좀 그려줘"
```

```python
#3.9.1 subplots _ countplots
cat_columns = ['Survived','Pclass','Sex']

##nrows 1이고 ncols는 칼럼의 개수만큼은 subplots을 설정
fig,axs=plt.subplots(nrows=1, ncols=len(cat_columns), figsize=(14,4))

for index, column in enumerate(cat_columns):
    print('index:' , index)
    #seaborn의 Axes 레벨 function들은 ax인자로 sublots의 어느 Axes에 위치할지 설정
    sns.countplot(x=column, data=titanic_df, ax=axs[index])
```

```python
#3.9.2 subplots _ barplots
cat_columns = ['Pclass','Sex','Embarked']

##nrows 1이고 ncols는 칼럼의 개수만큼은 subplots을 설정
fig,axs=plt.subplots(nrows=1, ncols=len(cat_columns), figsize=(14,4))

for index, column in enumerate(cat_columns):
    #seaborn의 Axes 레벨 function들은 ax인자로 sublots의 어느 Axes에 위치할지 설정
    sns.barplot(x=column, y='Survived', data=titanic_df, ax=axs[index])
```

```python
#3.9.3 subplots _ violinplots
cont_columns = ['Age','Fare']

# 리스트로 할당된 칼럼들의 개수만큼 루프 수행
for column in cont_columns:
    #왼쪽에는 바이올린 플롯, 오른쪽에는 히스토그램을 시각화, nrows는 1, ncols=2인 서브플롯 생성
    fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(10,4))
    #왼쪽 Axes 객체에는 Survived값 0,1별 개별 칼럼의 바이올린 플롯 시각화
    sns.violinplot(x='Survived', y=column, data=titanic_df, ax=axs[0])
    #오른쪽 Axes 객체이는 Survived 값에 따른 개별 칼럼의 히스토그램 시각화
    sns.histplot(x=column, data=titanic_df, kde=True, hue='Survived', ax=axs[1])
```
