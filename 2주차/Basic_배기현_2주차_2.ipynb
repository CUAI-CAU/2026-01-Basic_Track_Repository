### Matplotlib plot() 함수와 그래프 스타일 설정

`plot()` 함수는 x좌표와 y좌표 데이터를 입력받아 선 그래프를 생성한다.  
그래프의 입력 데이터는 **파이썬 리스트(list)** 또는 **NumPy 배열(array)** 모두 사용할 수 있다.

그래프를 그릴 때 x좌표와 y좌표 데이터의 길이는 반드시 동일해야 한다.

`plot()` 함수는 다양한 옵션을 통해 그래프의 스타일을 설정할 수 있다.

- `color` : 그래프 선의 색상 설정  
- `marker` : 데이터 포인트 표시  
- `linestyle` : 선의 스타일 설정  
- `linewidth` : 선의 두께 설정  
- `markersize` : 마커의 크기 설정

이러한 옵션을 활용하면 그래프의 색상, 선 모양, 데이터 포인트 표시 등을 다양하게 변경할 수 있다.

```python
plt.plot(x_value, y_value, color='red', marker='o', linestyle='dashed', linewidth=2, markersize=12)

# x축과 y축의 이름을 텍스트로 할당하여 설정. 
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.show()
```

```python
# x축, y축 틱값을 표현을 회전해서 보여줌. x축값이 문자열이고 많은 tick값이 있을 때 적용. 

# x 좌표 값을 0 ~ 99까지 100개로 설정. 
x_value = np.arange(0, 100)
# y 좌표 값은 개별 x좌표값을 2배로 해서 100개로 설정. 
y_value = 2*x_value

plt.plot(x_value, y_value, color='green')
plt.xlabel('x axis')
plt.ylabel('y axis')

# x축 눈금값을 90도로 회전
plt.xticks(rotation=90)

plt.show()
```

### Matplotlib 축 레이블과 눈금(ticks) 설정

`xlabel()`과 `ylabel()` 함수를 사용하면 그래프의 x축과 y축에 이름을 지정할 수 있다.  
그래프를 해석하기 쉽게 하기 위해 축의 의미를 설명할 때 사용된다.

또한 `xticks()` 함수를 이용하면 x축의 눈금(tick) 값을 회전하거나 원하는 값으로 지정할 수 있다.  
특히 문자열 데이터가 많거나 눈금이 겹칠 때 회전하여 표시하면 가독성을 높일 수 있다.

`rotation` 옵션은 눈금 값을 회전시키는 역할을 하며, 예를 들어 `rotation=90`은 눈금을 90도 회전시켜 세로로 표시한다.

또한 `ticks` 옵션을 사용하면 표시할 눈금 값을 직접 지정할 수 있다.  
예를 들어 `np.arange(0, 100, 5)`를 사용하면 0부터 99까지 5 간격으로 눈금을 표시할 수 있다.

마지막으로 `title()` 함수는 그래프의 제목을 설정하고,  
`show()` 함수는 생성된 그래프를 화면에 출력한다.

```python
# X좌표값은 0 ~ 99까지 100개, 첫번째 Y좌표값은 X좌표값의 두배, 두번째 좌표값은
# X좌표값의 4배
x_value_01 = np.arange(0, 100)
y_value_01 = 2*x_value_01
y_value_02 = 4*x_value_01

# X좌표값과 첫번째 Y좌표값으로 녹색의 선그래프를 그림. label을 'green line'으로 설정.  
plt.plot(x_value_01, y_value_01, color='green', label='green line')
# X좌표값과 두번째 Y좌표값으로 빨간색 선그래프를 그림. label을 'red line'으로 설정. 
plt.plot(x_value_01, y_value_02, color='red', label='red line')

plt.xlabel('x axis')
plt.ylabel('y axis')
# 개별 plot들의 label로 설정된 문자열을 범례로 표시
plt.legend()

plt.title('Hello plot')

plt.show()
```

### Matplotlib 범례(legend) 표시

여러 개의 그래프를 하나의 Figure에 함께 표시할 때 `label` 옵션을 사용하면  
각 그래프를 구분할 수 있는 이름을 지정할 수 있다.

`plot()` 함수의 `label` 파라미터에 문자열을 지정하면 해당 그래프의 이름이 설정되며,  
`legend()` 함수를 호출하면 이 label 값들이 그래프의 **범례(legend)** 로 표시된다.

또한 `xlabel()`과 `ylabel()`을 사용하여 x축과 y축의 이름을 설정할 수 있으며,  
`title()` 함수를 통해 그래프의 제목을 지정할 수 있다.

이 코드에서는 두 개의 선 그래프를 생성한다.

- 첫 번째 그래프 : `y = 2x` (green line)
- 두 번째 그래프 : `y = 4x` (red line)

`legend()` 함수를 사용하여 두 그래프의 label이 범례로 표시되도록 설정한다.

```python
x_value_01 = np.arange(1, 10)
y_value_01 = 2*x_value_01

# 마커를 포함한 빨간선 대쉬 선 그래프를 그리고, label은 'red line'으로 설정  
plt.plot(x_value_01, y_value_01, color='red', marker='o', linestyle='dashed', label='red line')
# X값에 따른 Y값을 나타내는 초록색 막대 그래프를 그리고 label은 'bar plot'으로 설정
plt.bar(x_value_01, y_value_01, color='green', label='bar plot')
plt.xlabel('x axis')
plt.ylabel('y axis')
# 두개의 그래프에서 label설정된 문자열 값을 범례로 표시
plt.legend()

plt.title('Hello plot')

plt.show()
```

```python
figure = plt.figure()
# Axes 객체를 추출하여 ax변수에 할당. 
ax = plt.axes()

# plt.plot( )은 ax.plot( )으로 plt.bar( )는 ax.bar( )로 변경. 
ax.plot(x_value_01, y_value_01, color='red',  marker='o', linestyle='dashed', label='red line')
ax.bar(x_value_01, y_value_01, color='green', label='bar plot')

# plt.xlabel( )은 ax.set_xlabel( )로, plt.ylabel( )은 ax.set_ylabel( )로 변경
ax.set_xlabel('x axis')
ax.set_ylabel('y axis')

# ax.set_legend()가 아니라 ax.legend()임. 
ax.legend()
# plt.title( )을 plt.set_title( )로 변경
ax.set_title('Hello plot')

plt.show()
```

### Matplotlib 선 그래프와 막대 그래프 함께 표현

Matplotlib에서는 `plot()` 함수를 이용해 선 그래프를, `bar()` 함수를 이용해 막대 그래프를 그릴 수 있다.  
여러 개의 그래프를 하나의 Figure에 함께 표시할 경우 `label` 옵션을 사용하여 그래프의 이름을 지정하고,

또한 Matplotlib은 **객체지향 방식(Object-Oriented 방식)**으로도 그래프를 생성할 수 있다.

### subplots()로 생성된 Axes 객체 타입 확인

`subplots()` 함수로 여러 개의 그래프 영역을 생성하면  
Axes 객체는 **배열 형태(numpy.ndarray)** 로 반환된다.

따라서 `ax` 변수는 여러 개의 Axes 객체를 담고 있는 배열이며,  
`ax[0]`, `ax[1]`과 같이 인덱스를 이용해 각각의 그래프 영역에 접근할 수 있다.

`type()` 함수를 사용하면 `ax`, `ax[0]`, `ax[1]`의 객체 타입을 확인할 수 있으며,  
각 인덱스에 해당하는 요소는 개별 **Axes 객체**임을 확인할 수 있다.

```python
import numpy as np

x_value_01 = np.arange(1, 10)
x_value_02 = np.arange(1, 20)
y_value_01 = 2 * x_value_01
y_value_02 = 2 * x_value_02

# figsize는 (6,3)인 Figure와 2개의 Axes를 가지는 subplots을 반환. 
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(6, 3))

# 개별 Axes객체들을 각각 이용하여 선 그래프와 막대그래프를 그림. 
ax[0].plot(x_value_01, y_value_01, color='red', marker='o', linestyle='dashed', label='red line')
ax[1].bar(x_value_02, y_value_02, color='green', label='bar plot')

# 개별 Axes객체 각각에 X축명을 설정. 
ax[0].set_xlabel('ax[0] x axis')
ax[1].set_xlabel('ax[1] x axis')

# 개별 Axes객체 각각에 범례를 설정.
ax[0].legend()
ax[1].legend()

# 개별 Axes객체 각각에 타이틀을 설정.
ax[0].set_title('Hello line')
ax[1].set_title('Hello bar')

plt.show()
```

### subplots()를 이용한 여러 개의 그래프 생성 및 설정

`subplots()` 함수는 하나의 Figure 안에 여러 개의 Axes(그래프 영역)를 생성할 때 사용된다.  
`nrows`와 `ncols` 옵션을 이용해 그래프의 행과 열 개수를 지정할 수 있으며, `figsize`로 Figure의 크기를 설정할 수 있다.

여러 개의 Axes가 생성되면 `ax[0]`, `ax[1]`과 같이 인덱스를 이용해 각각의 그래프 영역에 접근할 수 있다.

각 Axes 객체에 대해 `plot()`이나 `bar()` 함수를 사용하여 서로 다른 그래프를 그릴 수 있으며,  
`set_xlabel()`, `set_title()` 등을 사용해 축 이름과 제목을 개별적으로 설정할 수 있다.

또한 `legend()` 함수를 사용하면 각 그래프의 label 값을 범례로 표시할 수 있다.

이 코드에서는 두 개의 Axes를 생성하여  
첫 번째 그래프에는 선 그래프를, 두 번째 그래프에는 막대 그래프를 각각 표현한다.

```python
import numpy as np

x_value_01 = np.arange(1, 10)
x_value_02 = np.arange(1, 20)
y_value_01 = 2 * x_value_01
y_value_02 = 2 * x_value_02

# figsize는 (6,6)인 Figure와 2x2개의 Axes를 가지는 subplots을 반환.
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(6, 6))

# 2x2 Axes객체들을 각각 이용하여 선 그래프와 막대그래프를 그림. 
ax[0][0].plot(x_value_01, y_value_01, color='red', marker='o', linestyle='dashed', label='red line')
ax[0][1].bar(x_value_02, y_value_02, color='green', label='green bar')
ax[1][0].plot(x_value_01, y_value_01, color='green', marker='o', linestyle='dashed', label='green line')
ax[1][1].bar(x_value_02, y_value_02, color='red', label='red bar')

# 개별 Axes객체 각각에 X축명을 설정. 
ax[0][0].set_xlabel('ax[0][0] x axis')
ax[0][1].set_xlabel('ax[0][1] x axis')
ax[1][0].set_xlabel('ax[1][0] x axis')
ax[1][1].set_xlabel('ax[1][1] x axis')

# 개별 Axes객체 각각에 범례를 설정 
ax[0][0].legend()
ax[0][1].legend() 
ax[1][0].legend()
ax[1][1].legend() 

plt.show()
```

### subplots()를 이용한 2x2 형태의 여러 그래프 생성

`subplots()` 함수는 하나의 Figure 안에 여러 개의 Axes(그래프 영역)를 생성할 수 있다.  
`nrows`와 `ncols` 옵션을 사용하면 그래프의 행과 열 구조를 지정할 수 있으며, `figsize`로 Figure의 크기를 설정한다.

위 코드는 **2행 2열(2x2)** 형태의 Axes를 생성하며, `ax[행][열]` 형태의 인덱스를 이용해 각각의 그래프 영역에 접근할 수 있다.

각 Axes 객체에 대해 `plot()` 또는 `bar()` 함수를 사용하여 서로 다른 그래프를 그릴 수 있으며,

이 코드에서는 네 개의 Axes에 각각 **선 그래프와 막대 그래프**를 배치하여 여러 그래프를 한 Figure 안에서 동시에 표현한다.

### Seaborn을 이용한 히스토그램 시각화

Seaborn은 matplotlib 기반의 데이터 시각화 라이브러리로, 통계 기반 그래프를 쉽게 표현할 수 있다.

`titanic_df` 데이터셋을 `pd.read_csv()`로 불러온 뒤 `head()` 함수를 사용하면  
데이터의 상위 몇 개 행을 확인할 수 있다.

`sns.histplot()` 함수는 데이터의 분포를 확인할 때 사용하는 **히스토그램(histogram)** 을 생성한다.  
히스토그램은 특정 값이 얼마나 자주 나타나는지를 막대 형태로 표현한다.

`bins` 옵션은 데이터를 몇 개의 구간으로 나누어 표시할지를 설정한다.

`x` 파라미터에 컬럼명을 지정하고 `data`에 DataFrame을 입력하면

### Seaborn violinplot을 이용한 데이터 분포 시각화

`sns.violinplot()` 함수는 데이터의 분포를 **확률 밀도 곡선과 박스플롯 형태를 함께 사용하여 시각화**하는 그래프이다.

바이올린 플롯은 데이터의 분포 형태를 확인하기 좋으며, 범주별 데이터 분포 비교에 자주 사용된다.

`y`에 컬럼을 지정하면 해당 데이터의 전체 분포를 확인할 수 있고,  
`x`에 범주형 변수를 지정하면 범주별 데이터 분포를 비교할 수 있다.

또한 `hue` 파라미터를 사용하면 추가적인 범주 기준으로 데이터를 색상별로 구분하여 비교할 수 있다.

이 코드에서는

- `Age` 컬럼의 전체 분포를 바이올린 플롯으로 시각화  
- `Pclass` 기준으로 승객 나이 분포 비교  
- `hue='Sex'`를 사용하여 객실 등급 내에서 성별에 따른 나이 분포를 함께 비교

하는 그래프를 생성한다.

```python
cat_columns = ['Survived', 'Pclass', 'Sex']

# nrows는 1이고 ncols는 컬럼의 갯수만큼인 subplots을 설정. 
fig, axs = plt.subplots(nrows=1, ncols=len(cat_columns), figsize=(14, 4))

for index, column in enumerate(cat_columns):
    print('index:', index)
    # seaborn의 Axes 레벨 function들은 ax인자로 subplots의 어느 Axes에 위치할지 설정. 
    sns.countplot(x=column, data=titanic_df, ax=axs[index])
```

```python
cat_columns = ['Pclass', 'Sex', 'Embarked']

# nrows는 1이고 ncols는 컬럼의 갯수만큼인 subplots을 설정. 
fig, axs = plt.subplots(nrows=1, ncols=len(cat_columns), figsize=(14, 4))

for index, column in enumerate(cat_columns):
    # seaborn의 Axes 레벨 function들은 ax인자로 subplots의 어느 Axes에 위치할지 설정. 
    sns.barplot(x=column, y='Survived', data=titanic_df, ax=axs[index])
```

```python
cont_columns = ['Age', 'Fare']

# 리스트로 할당된 컬럼들의 개수만큼 루프 수행.
for column in cont_columns:
    # 왼쪽에는 바이올린 플롯, 오른쪽에는 히스토그램을 시각화 하기 위해 nrows는 1, ncols=2인 서브플롯 생성. 
    fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))
    # 왼쪽 Axes객체에는 Survived값 0, 1별 개별 컬럼의 바이올린 플롯 시각화.
    sns.violinplot(x='Survived', y=column, data=titanic_df, ax=axs[0])
    # 오른쪽 Axes객체에는 Survived 값에 따른 개별 컬럼의 히스토그램 시각화 
    sns.histplot(x=column, data=titanic_df, kde=True, hue='Survived', ax=axs[1])
```

### 여러 변수에 대한 반복 시각화

여러 개의 컬럼을 한 번에 시각화하기 위해 리스트와 반복문(`for`)을 활용할 수 있다.  
`subplots()`을 이용해 여러 개의 Axes를 생성한 뒤, Seaborn 그래프 함수의 `ax` 파라미터에  
각 Axes 객체를 지정하면 원하는 위치에 그래프를 배치할 수 있다.

첫 번째 예제에서는 `Survived`, `Pclass`, `Sex`와 같은 **범주형 변수**에 대해  
`countplot()`을 이용하여 각 범주의 데이터 개수를 시각화한다.

두 번째 예제에서는 `Pclass`, `Sex`, `Embarked`와 같은 범주형 변수와  
`Survived`를 함께 사용하여 `barplot()`으로 **범주별 생존율 평균**을 비교한다.

마지막 예제에서는 `Age`, `Fare`와 같은 **연속형 변수**에 대해

### Seaborn scatterplot을 이용한 두 변수 관계 시각화

`sns.scatterplot()` 함수는 두 개의 연속형 변수 간의 관계를 점(point) 형태로 시각화할 때 사용된다.

`x`와 `y` 파라미터에 컬럼을 지정하면 두 변수의 값 관계를 산점도로 표현할 수 있다.

또한 `hue` 옵션을 사용하면 특정 범주형 변수 기준으로 점의 색상을 구분할 수 있으며,  
`style` 옵션을 사용하면 범주에 따라 점의 모양을 다르게 표시할 수 있다.

이 코드에서는

- `Age`와 `Fare`의 관계를 산점도로 시각화하고  
- `hue='Survived'`를 사용하여 생존 여부에 따라 색상을 구분하며  
- `style='Sex'`를 사용하여 성별에 따라 점의 모양을 구분한다.
