```python
#fillna로 결손데이터 대체
#cabin 칼럼의 nan값을 c000으로 대체
titanic_df['Cabin']=titanic_df['Cabin'].fillna('C000')
titanic_df.head(3)
```

```python
#결손데이터를 평균 또는 대체값으로 처리
titanic_df['Age']=titanic_df['Age'].fillna(titanic_df['Age'].mean())
titanic_df['Embarked']=titanic_df['Embarked'].fillna('S')
titanic_df.isna().sum()
```

```python
#lambda식을 통한 데이터 가공
a=[1,2,3]
squares=map(lambda x:x**2,a)
list(squares)
```

```python
#name_len이라는 칼럼을 만들고 거기에 name의 길이를 저장
titanic_df['Name_len']=titanic_df['Name'].apply(lambda x:len(x))
titanic_df[['Name','Name_len']].head(3)
```

```python
#조건문을 활용한 lambda
#나이가 15세 미만이면 child, 아니면 adult로 분류하고 child_adult 칼럼에 저장
titanic_df['Child_Adult']=titanic_df['Age'].apply(lambda x: 'Child' if x<=15 else 'Adult')
titanic_df[['Age','Child_Adult']].head(8)
```

```python
#if, else는 지원하지만 else if는 지원하지않아서 else절 안에서 ()로 if else 적용
titanic_df['Age_cat']=titanic_df['Age'].apply(lambda x: 'Child' if x<=15 else ('Adult' if x<=60 else 'Elderly'))
titanic_df['Age_cat'].value_counts()
```

```python
#else if가 많이 필요한경우 별도의 함수를 만들어 분류
#나이에 따라 세분화된 분류를 수행하는 함수 생성
def get_category(age):
    cat=''
    if age<=5:
        cat='Baby'
    elif age<=12:
        cat='Child'
    elif age<=18:
        cat='Teenager'
    elif age<=25:
        cat='Student'
    elif age<=35:
        cat='Young Adult'
    elif age<=60:
        cat='Adult'
    else:
        cat='Elderly'
    return cat

#lambda 식에 위에서 생성한 get_category 함수를 반환값으로 지정
titanic_df['Age_cat']=titanic_df['Age'].apply(lambda x:get_category(x))
titanic_df[['Age','Age_cat']].head()
```
