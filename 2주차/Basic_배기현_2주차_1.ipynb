### 리스트와 NumPy 배열을 이용한 DataFrame 생성

pandas의 `DataFrame`은 다양한 형태의 데이터를 이용해 생성할 수 있다.  
대표적으로 파이썬의 리스트(list)나 NumPy 배열(ndarray)을 이용해 DataFrame을 만들 수 있다.

- `pd.DataFrame(data, columns=[])` : 데이터를 DataFrame 형태로 생성
- `columns` : DataFrame의 컬럼 이름 지정
- 리스트(list)와 NumPy 배열(ndarray)은 모두 DataFrame 생성에 활용 가능

이를 통해 다양한 형태의 데이터를 pandas의 표 형태 데이터 구조로 변환하여 데이터 분석에 활용할 수 있다.

```python
#3개의 칼럼명이 필요함.
col_name2=['col1','col2','col3']

#2행x3열 형태의 리스트와 ndarray 생성한 뒤 이를 DataFrame으로 변환.
list2 = [[1,2,3],
         [11,12,13]]
array2 = np.array(list2)
print('array2 shape:', array2.shape)
df_list2 = pd.DataFrame(list2, columns=col_name2)
print('2차원 리스트로 만든 DataFrame:\n',df_list2)
df_array2 = pd.DataFrame(array2, columns=col_name2)
print('2차원 array로 만든 DataFrame:\n',df_array2)
```

### 2차원 리스트와 NumPy 배열을 이용한 DataFrame 생성

pandas의 `DataFrame`은 2차원 형태의 데이터를 이용하여 생성할 수 있다.  
2차원 리스트(list)나 NumPy 배열(ndarray)을 사용하면 여러 개의 컬럼을 가진 DataFrame을 만들 수 있다.

- `pd.DataFrame(data, columns=[])` : 데이터를 DataFrame으로 변환
- `columns` : DataFrame의 컬럼 이름 지정
- 2차원 데이터 구조는 `(행, 열)` 형태로 구성된다.

이 예제에서는 2행 3열 구조의 리스트와 NumPy 배열을 이용하여 DataFrame을 생성한다.

```python
#Key는 문자열 칼럼명으로 매핑, Value는 리스트 형(또는 ndarray) 칼럼 데이터로 매핑
dict = {'col1':[1,11],'col2':[2,22],'col3':[3,33]}
df_dict = pd.DataFrame(dict)
print('딕셔너리로 만든 DataFrame:\n', df_dict)
```

### 딕셔너리를 이용한 DataFrame 생성

pandas의 `DataFrame`은 딕셔너리(dictionary) 구조를 이용해 생성할 수 있다.  
이때 딕셔너리의 **Key는 컬럼 이름(column name)**이 되고,  
**Value는 해당 컬럼에 들어갈 데이터(리스트 또는 NumPy 배열)**가 된다.

- `Key` → DataFrame의 컬럼 이름
- `Value` → 컬럼에 해당하는 데이터

이 방식은 각 컬럼 데이터를 직접 지정할 수 있기 때문에  
pandas에서 DataFrame을 생성할 때 가장 많이 사용되는 방법이다.

```python
#DataFrame을 ndarray로 변환
array3 = df_dict.values
print('df_dict.values 타입:',type(array3),'df_dict.values shape:',array3.shape)
print(array3)
```

### DataFrame을 NumPy 배열(ndarray)로 변환

pandas DataFrame은 `.values` 속성을 이용하여 NumPy 배열(ndarray) 형태로 변환할 수 있다.  
NumPy 배열은 행렬 연산과 수치 계산에 효율적인 데이터 구조로, 머신러닝이나 수치 계산에서 많이 사용된다.

- `DataFrame.values` : DataFrame 데이터를 NumPy ndarray 형태로 변환
- `type()` : 데이터 타입 확인
- `shape` : 배열의 형태(행, 열) 확인

이 과정을 통해 pandas 데이터 구조를 NumPy 배열로 변환하여 다양한 수치 연산에 활용할 수 있다.

```python
#DataFrame을 리스트로 변환
list3 = df_dict.values.tolist()
print('df_dict.values.tolist 타입:',type(list3))
print(list3)
#DataFrame을 딕셔너리로 변환
dict3 = df_dict.to_dict('list')
print('\n df_dict.to_dict 타입:',type(dict3))
print(dict3)
```

### DataFrame을 리스트와 딕셔너리로 변환

pandas DataFrame은 다양한 파이썬 자료형으로 변환할 수 있다.  
대표적으로 NumPy 배열을 거쳐 리스트(list)로 변환하거나, 딕셔너리(dictionary) 형태로 변환할 수 있다.

- `DataFrame.values.tolist()` : DataFrame 데이터를 리스트 형태로 변환
- `DataFrame.to_dict()` : DataFrame 데이터를 딕셔너리 형태로 변환
- `to_dict('list')` : 각 컬럼 데이터를 리스트 형태로 변환하여 딕셔너리로 반환

이를 통해 pandas 데이터를 다른 파이썬 자료구조로 변환하여 다양한 방식으로 활용할 수 있다.

### DataFrame에서 컬럼 삭제 (drop)

pandas에서는 `drop()` 함수를 사용하여 특정 행(row)이나 열(column)을 삭제할 수 있다.

- `drop('컬럼명', axis=1)` : 특정 컬럼 삭제
- `axis=1` : 열(column) 방향 삭제
- `axis=0` : 행(row) 방향 삭제
- `inplace=True` : 삭제 결과를 기존 DataFrame에 바로 반영

기본적으로 `drop()` 함수는 새로운 DataFrame을 반환하며 원본 데이터는 변경되지 않는다.  
하지만 `inplace=True` 옵션을 사용하면 기존 DataFrame을 직접 수정할 수 있다.

```python
pd.set_option('display.width', 1000)
pd.set_option('display.max_colwidth',15)
print('#### before axis 0 drop ####')
print(titanic_df.head(3))

titanic_df.drop([0,1,2], axis=0, inplace=True)

print('#### after axis 0 drop ####')
print(titanic_df.head(3))
```

### Index 객체의 특징과 Series 연산

pandas DataFrame의 Index는 각 행(row)을 식별하는 역할을 하며,  
NumPy 배열과 유사한 구조를 가진다. 하지만 Index 객체는 **immutable(변경 불가능)**한 특징을 가진다.

- `Index.values` : Index 값을 NumPy 배열 형태로 반환
- `Index[index]` : 특정 위치의 Index 값 조회

또한 pandas의 `Series` 객체는 다양한 통계 연산과 벡터 연산을 지원한다.

- `Series.max()` : 최대값 계산
- `Series.sum()` : 합계 계산
- `Series + 값` : 모든 요소에 대해 벡터 연산 수행

이러한 기능을 통해 pandas Series는 효율적인 데이터 분석과 수치 계산을 수행할 수 있다.

### DataFrame 데이터 선택 (Indexing)

pandas DataFrame에서는 `[]` 연산자를 사용하여 특정 컬럼이나 데이터의 일부를 선택할 수 있다.

- `DataFrame['컬럼명']` : 하나의 컬럼 선택 (Series 반환)
- `DataFrame[['컬럼1','컬럼2']]` : 여러 컬럼 선택 (DataFrame 반환)
- `DataFrame[start:end]` : 행(row) 슬라이싱
- `DataFrame[조건]` : 조건에 맞는 행 선택 (Boolean Indexing)

이를 통해 필요한 데이터만 선택하여 분석을 수행할 수 있다.

### iloc 인덱싱 (위치 기반 인덱싱)

pandas에서 `iloc`은 **정수 위치(integer position)를 기반으로 데이터를 선택하는 방법**이다.

- `iloc[row_index, column_index]`
- 행(row)과 열(column)의 **정수 위치 값만 사용 가능**
- 문자열 컬럼명이나 인덱스 이름은 사용할 수 없음

따라서 `iloc`은 배열과 유사하게 **0부터 시작하는 위치 기반 인덱싱**을 사용한다.

예를 들어 `iloc[0,0]`은 첫 번째 행과 첫 번째 열의 데이터를 의미한다.

```python
print("\n 맨 마지막 칼럼 데이터 [:, -1] \n", data_df.iloc[:, -1])
print("\n 맨 마지막 칼럼을 제외한 모든 데이터 [:, :-1] \n", data_df.iloc[:, :-1])
```

### iloc을 이용한 슬라이싱(Slicing)

`iloc`은 정수 위치 기반 인덱싱을 사용하며, 슬라이싱(slice)을 통해 여러 행이나 열을 선택할 수 있다.

- `:` : 모든 행 또는 열 선택
- `-1` : 마지막 위치를 의미
- `:-1` : 마지막을 제외한 모든 데이터 선택

`iloc` 슬라이싱을 이용하면 특정 범위의 데이터를 효율적으로 추출할 수 있다.

```python
data_df.loc['one','Name']
```

```python
data_df.loc[0, 'Name']
```

```python
print('위치기반 iloc slicing\n', data_df.iloc[0:1, 0], '\n')
print('명칭기반 loc slicing\n', data_df.loc['one':'two','Name'])
```

### GroupBy를 이용한 데이터 집계

pandas에서는 `groupby()` 함수를 사용하여 특정 컬럼을 기준으로 데이터를 그룹화할 수 있다.

`groupby()`는 지정한 컬럼의 값에 따라 데이터를 그룹으로 나눈 후,  
각 그룹에 대해 다양한 집계 함수(aggregation function)를 적용할 수 있다.

대표적으로 다음과 같은 집계 함수를 사용할 수 있다.

- `count()` : 데이터 개수 계산
- `sum()` : 합계 계산
- `mean()` : 평균 계산
- `max()` : 최대값 계산
- `min()` : 최소값 계산

또한 특정 컬럼만 선택하여 그룹별 통계 정보를 계산할 수도 있다.

```python
titanic_df.groupby('Pclass')['Age'].agg([max,min])
```

```python
agg_format={'Age':'max','SibSp':'sum','Fare':'mean'}
titanic_df.groupby('Pclass').agg(agg_format)
```

이를 통해 그룹별 데이터의 다양한 통계 정보를 효율적으로 분석할 수 있다.

### 결측값(Missing Value) 처리

pandas에서는 데이터에 존재하는 결측값(NaN)을 확인하고 처리하기 위한 다양한 함수를 제공한다.

- `isna()` : 데이터가 결측값인지 여부를 확인
- `sum()` : 각 컬럼의 결측값 개수 계산
- `fillna()` : 결측값을 특정 값으로 대체

결측값은 데이터 분석이나 머신러닝 모델 학습 과정에서 오류를 발생시킬 수 있으므로  
적절한 값으로 대체하거나 제거하는 전처리 과정이 필요하다.

예를 들어 문자열 컬럼은 특정 문자열로 대체하고,  
수치형 데이터는 평균값이나 중앙값으로 결측값을 채우는 방식이 자주 사용된다.

```python
def get_sqaure(a):
    return a**2

print('3의 제곱은:', get_sqaure(3))
```

```python
lambda_square = lambda x:x**2
print('3의 제곱은:', lambda_square)
```

```python
a=[1,2,3]
squares = map(lambda x:x**2,a)
list(squares)
```

### Lambda 함수와 map()

Python에서는 간단한 함수를 정의할 때 `lambda` 함수를 사용할 수 있다.  
`lambda` 함수는 이름 없이 한 줄로 작성할 수 있는 익명 함수(anonymous function)이다.

기본적인 함수 정의 방식은 `def` 키워드를 사용하지만,  
간단한 연산의 경우 `lambda`를 이용하면 코드가 더 간결해진다.

또한 `map()` 함수와 함께 사용하면 리스트의 각 요소에 동일한 함수를 적용할 수 있다.

- `lambda x : x**2` : 입력값 x의 제곱 계산
- `map(function, iterable)` : iterable의 각 요소에 함수 적용


```python
titanic_df['Name_len']=titanic_df['Name'].apply(lambda x : len(x))
titanic_df[['Name','Name_len']].head(3)
```

```python
titanic_df['Child_Adult']=titanic_df['Age'].apply(lambda x : 'Child' if x <= 15 else 'Adult')
titanic_df[['Age','Child_Adult']].head(8)
```

```python
titanic_df['Age_cat']=titanic_df['Age'].apply(lambda x : 'Child' if x <= 15 else ('Adult' if x<=60 else 'Elderly'))
titanic_df['Age_cat'].value_counts()
```

```python
def get_category(age):
    cat = ''
    if age <= 5: cat = 'Baby'
    elif age <= 12: cat = 'Child'
    elif age <= 18: cat = 'Teenager'
    elif age <= 25: cat = 'Student'
    elif age <= 35: cat = 'Young Adult'
    elif age <= 60: cat = 'Adult'
    else: cat = 'Elderly'

    return cat

#lambda 식에 위에서 생성한 get_category() 함수를 반환값으로 지정.
#get_category(X)는 입력값으로 'Age' 칼럼 값을 받아서 해당하는 cat 반환
titanic_df['Age_cat'] = titanic_df['Age'].apply(lambda x :get_category(x))
titanic_df[['Age','Age_cat']].head()
```

### apply()와 lambda를 이용한 Feature Engineering

pandas에서는 `apply()` 함수와 `lambda` 함수를 이용해 기존 데이터를 기반으로 새로운 컬럼을 생성할 수 있다.

`apply()` 함수는 DataFrame이나 Series의 각 요소에 대해 특정 함수를 적용할 때 사용되며,  
`lambda` 함수를 함께 사용하면 간단한 데이터 변환을 한 줄로 표현할 수 있다.

예를 들어 문자열 길이를 계산하거나, 나이에 따라 연령대를 구분하는 새로운 컬럼을 생성할 수 있다.

또한 복잡한 조건이 필요한 경우 별도의 함수를 정의한 뒤 `apply()`에 적용할 수도 있다.

이러한 과정은 머신러닝에서 **Feature Engineering(특성 생성)** 단계로 매우 중요하다.
