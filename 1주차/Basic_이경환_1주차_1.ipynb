### Numpy ndarray 개요

```python
import numpy as np
#넘파이를 가져와줍시다.
```

```python
array1 = np.array([1,2,3])
print('array1 type:',type(array1))
print('array1 array 형태:',array1.shape)
#array1이 길이가 3인 1차원 배열임을 확인합니다.
array2 = np.array([[1,2,3],
                  [2,3,4]])
print('array2 type:',type(array2))
print('array2 array 형태:',array2.shape)
#array2가 크기가 2,3인 2차원 배열임을 확인합니다.
array3 = np.array([[1,2,3]])
print('array3 type:',type(array3))
print('array3 array 형태:',array3.shape)
#array3이 크기가 1,3인 2차원 배열임을 확인합니다.
```

```python
print('array1: {0}차원, array2: {1}차원, array3: {2}차원'.format(array1.ndim,array2.ndim,array3.ndim))
#차원을 확인합니다.
```

```python
list1 = [1,2,3]
print(type(list1))
array1 = np.array(list1)
print(type(array1))
print(array1, array1.dtype)
#데이터 타입을 확인합니다. 
#print(array1, list1.dtype)
```

```python
list2 = [1, 2, 'test']
array2 = np.array(list2)
print(array2, array2.dtype)

list3 = [1, 2, 3.0]
array3 = np.array(list3)
print(array3, array3.dtype)
#리스트의 데이터중 하나를 실수나 문자열로 바꾸면 나머지 숫자들도 실수, 문자로 바뀌는 현상도 관찰됩니다.
```

```python
array_int = np.array([1, 2, 3])
array_float = array_int.astype('float64')
print(array_float, array_float.dtype)

array_int1= array_float.astype('int32')
print(array_int1, array_int1.dtype)

array_float1 = np.array([1.1, 2.1, 3.1])
array_int2= array_float1.astype('int32')
print(array_int2, array_int2.dtype)
```

### reshape

```python
array1 = np.arange(10)
print('array1:\n', array1)

array2 = array1.reshape(2,5)
print('array2:\n',array2)

array3 = array1.reshape(5,2)
print('array3:\n',array3)
#같은 데이터를 다양한 차원에 넣을 수 있습니다.
```

```python
array1.reshape(4,3)
#하지만 데이터 크기와 같은 사이즈가 필요합니다.
#array1.reshape(4,2)
#사이즈가 작아도 안됩니다.
```

```python
array1 = np.arange(10)
print(array1)

array2 = array1.reshape(-1,5)
print('array2 shape:',array2.shape)
print(array2)

array3 = array1.reshape(5,-1)
print('array3 shape:',array3.shape)
print(array3)
#-1은 데이터 갯수에 따라 알아서 형태를 맞추라는 것을 의미합니다.
```

```python
array1 = np.arange(10)
array4 = array1.reshape(-1,4)
#10은 자동으로 만들려고 해도 4로 나눠 떨어지지 않아서 오류가 납니다.
```

```python
array1 = np.arange(8)
array3d = array1.reshape((2,2,2))
print('array3d:\n',array3d.tolist())

# 3차원 ndarray를 2차원 ndarray로 변환
array5 = array3d.reshape(-1,1)
print('array5:\n',array5.tolist())
print('array5 shape:',array5.shape)

# 1차원 ndarray를 2차원 ndarray로 변환
array6 = array1.reshape(-1,1)
print('array6:\n',array6.tolist())
print('array6 shape:',array6.shape)
```

* Slicing

```python
array1 = np.arange(start=1, stop=10)
array3 = array1[0:3]
#print(array1)
print(array3)
print(type(array3))
#지속적으로 보고있지만 1차원 배열과 리스트는 문법이 크게 차이나지 않습니다.
```

```python
array1 = np.arange(start=1, stop=10)
print(array1)
array4 = array1[:3]
print(array4)

array5 = array1[3:]
print(array5)

array6 = array1[:]
print(array6)
#리스트와 완벽히 똑같이 작동합니다.
```

* fancy indexing

```python
array1d = np.arange(start=1, stop=10)
array2d = array1d.reshape(3,3)

array3 = array2d[[0,1], 2]
print('array2d[[0,1], 2] => ',array3.tolist())
#array3 = array2d[0:2, 2]
#위와 같이 표기해도 될것을 굳이 다르게 표기하는 문법입니다. 읽을수만 있다면 굳이 사용할 필요는 없어보입니다.
array4 = array2d[[0,1], 0:2]
print('array2d[[0,1], 0:2] => ',array4.tolist())

array5 = array2d[[0,1]]
print('array2d[[0,1]] => ',array5.tolist())
```

* Boolean indexing

```python
array1d > 5
#array1d[array1d > 5] 을 쓰면 조건에 해당하는 값울
#array1d > 5과 같이 쓰면 조건이 만족했는지 여부를 판단하는 배열을 만듭니다.
```

```python
boolean_indexes = np.array([False, False, False, False, False,  True,  True,  True,  True])
array3 = array1d[boolean_indexes]
print('불린 인덱스로 필터링 결과 :', array3)
```

### 행렬의 정렬 – sort( )와 argsort( )

* 행렬 정렬

```python
org_array = np.array([ 3, 1, 9, 5]) 
print('원본 행렬:', org_array)
# np.sort( )로 정렬 
sort_array1 = np.sort(org_array)         
print ('np.sort( ) 호출 후 반환된 정렬 행렬:', sort_array1) 
print('np.sort( ) 호출 후 원본 행렬:', org_array)
# ndarray.sort( )로 정렬
sort_array2 = org_array.sort()
print('org_array.sort( ) 호출 후 반환된 행렬:', sort_array2)
print('org_array.sort( ) 호출 후 원본 행렬:', org_array)

#np.sort는 원본을 복사해 새로운 array를 만든 후 그것을 정렬시키고 ndarray.sort는 원래 행렬을 정렬시킵니다.
```

```python
sort_array1_desc = np.sort(org_array)[::-1]
print ('내림차순으로 정렬:', sort_array1_desc) 
```

```python
array2d = np.array([[8, 12], 
                   [7, 1 ]])

sort_array2d_axis0 = np.sort(array2d, axis=0)
print('로우 방향으로 정렬:\n', sort_array2d_axis0)

sort_array2d_axis1 = np.sort(array2d, axis=1)
print('컬럼 방향으로 정렬:\n', sort_array2d_axis1)
```

* 정렬 행렬의 인덱스 반환

```python
org_array = np.array([ 3, 1, 9, 5]) 
sort_indices = np.argsort(org_array)
print(type(sort_indices))
print('행렬 정렬 시 원본 행렬의 인덱스:', sort_indices)
#데이터가 정렬 되었다면 현재 인덱스가 몇번 인덱스로 옮겨져야 하는지 나타냅니다. 그냥 정렬된 데이터를 하나 더 만들면 될 것 같다는 생각이 들기는 합니다.
```

```python
org_array = np.array([ 3, 1, 9, 5]) 
sort_indices_desc = np.argsort(org_array)[::-1]
print('행렬 내림차순 정렬 시 원본 행렬의 인덱스:', sort_indices_desc)
```

```python
import numpy as np

name_array = np.array(['John', 'Mike', 'Sarah', 'Kate', 'Samuel'])
score_array= np.array([78, 95, 84, 98, 88])

sort_indices_asc = np.argsort(score_array)
print('성적 오름차순 정렬 시 score_array의 인덱스:', sort_indices_asc)
print('성적 오름차순으로 name_array의 이름 출력:', name_array[sort_indices_asc])
#학생성적 예시로는 여전히 쓸모있는지 체감이 잘 안됩니다. 굳이 인덱스를 뽑느니 그냥 정렬된 데이터를 만드는게 나아보입니다.
#데이터 과학을 조금 더 공부하며 쓸모를 제대로 이해해야 할 것 같습니다.
```

### 선형대수 연산 – 행렬 내적과 전치 행렬 구하기

* 행렬 내적

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([[7, 8],
              [9, 10],
              [11, 12]])

dot_product = np.dot(A, B)
print('행렬 내적 결과:\n', dot_product)
#a의 1행 b의 1열 = 58
#a의 1행 b의 2열 = 64
#a의 2행 b의 1열 = 139
#a의 2행 b의 2열 = 154
```

* 전치 행렬

```python
A = np.array([[1, 2],
              [3, 4]])
#A = np.array([[1, 2,3],
  #            [3, 4,5]])
transpose_mat = np.transpose(A)
print('A의 전치 행렬:\n', transpose_mat)
#전치행렬을 구합니다. 행열 길이가 맞지 않아도 전치행렬은 정의되기에 잘 구해집니다.
```
