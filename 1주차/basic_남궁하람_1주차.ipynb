## 넘파이 ndarray 개요

```python
# numpy 라이브러리를 np라는 이름으로 불러오기 
import numpy as np 
```

```python
array1 = np.array([1,2,3]) # array의 인자로는 리스트가 들어감 -> 리스트는 대괄호로 표현됨
print('array1 type:',type(array1)) # array 1은 numpy 배열임 -> 그러니 numpy.ndarray
print('array1 array 형태:',array1.shape) # 3개의 원소가 있는 1차원 배열 -> (3,)

array2 = np.array([[1,2,3],
                  [2,3,4]])
print('array2 type:',type(array2)) # array 2는 numpy 배열임 -> 그러니 numpy.ndarray
print('array2 array 형태:',array2.shape) # 2개의 행과 3개의 열이 있는 2차원 배열 -> (2, 3)

array3 = np.array([[1,2,3]])
print('array3 type:',type(array3)) # array 3은 numpy 배열임 -> 그러니 numpy.ndarray
print('array3 array 형태:',array3.shape) # 1개의 행과 3개의 열이 있는 2차원 배열 -> (1, 3)
# 위의 array3과 array1은 들어있는 숫자의 개수는 같지만 엄연히 다른 형태의 배열이다.
# 이는 괄호 개수로도 확인 가능하다.
```

```python
print('array1: {0}차원, array2: {1}차원, array3: {2}차원'.format(array1.ndim,array2.ndim,array3.ndim))
# array2와 array3는 모두 2차원 배열이다.
# 하지만 array2는 행이 2개이고, array3는 행이 1개인 2차원 배열이다.
```

```python
list1 = [1,2,3] 
print(type(list1)) # list1은 리스트임 -> 그러니 list
array1 = np.array(list1) # 리스트를 인자로 넣어 numpy 배열로 변환
print(type(array1)) # array1은 numpy 배열임 -> 그러니 numpy.ndarray
print(array1, array1.dtype) # array1의 원소는 모두 정수이므로 dtype은 int
```

```python
list2 = [1, 2, 'test'] # <= 리스트는 서로 다른 자료형이 섞여있기 가능
array2 = np.array(list2) # 하지만, numpy 배열은 모두 같은 자료형이어야 함.
print(array2, array2.dtype) # array2의 원소는 모두 문자열로 변환되어 나온다!

list3 = [1, 2, 3.0] # <= 이번엔 정수 + 실수로 구성된 리스트
array3 = np.array(list3) #
print(array3, array3.dtype) # array3의 원소는 모두 실수로 변환되어 나온다!
```

```python
array_int = np.array([1, 2, 3])
array_float = array_int.astype('float64') # -> numpy의 타입변경은 astype() 메서드를 이용하기
print(array_float, array_float.dtype) # 정수 -> 실수로 변환되어 나온다!

array_int1= array_float.astype('int32')
print(array_int1, array_int1.dtype) # 실수 -> 정수로 변환되어 나온다!

array_float1 = np.array([1.1, 2.1, 3.1]) # 실수 배열
array_int2= array_float1.astype('int32')
print(array_int2, array_int2.dtype) # 실수 -> 정수로 변환되어 나온다!
```

## ndarray를 편리하게 생성하기 - arange, zeros, ones

```python
sequence_array = np.arange(10) # -> 0 포함 10 미만의 정수들로 이루어진 1차원 배열 생성
print(sequence_array)
print(sequence_array.dtype, sequence_array.shape)
```

```python
zero_array = np.zeros((3,2),dtype='int32') # 모든 원소가 0인 3행 2열의 정수형 배열 생성
print(zero_array)
print(zero_array.dtype, zero_array.shape)

one_array = np.ones((3,2)) # 모든 원소가 1인 3행 2열의 실수형 배열 생성 
print(one_array)
print(one_array.dtype, one_array.shape) #-> dtype을 지정하지 않으면 기본적으로 실수형이 된다.
```

## reshape()

```python
array1 = np.arange(10) # -> 0부터 9까지의 정수로 이루어진 1차원 배열 생성
print('array1:\n', array1)

array2 = array1.reshape(2,5) # -> 10개의 원소를 2행 5열로 변환
print('array2:\n',array2)

array3 = array1.reshape(5,2) # -> 10개의 원소를 5행 2열로 변환
print('array3:\n',array3)
```

```python
array1.reshape(4,3) # 오류가 나는 이유 -> 10개의 원소를 4행 3열로 변환하려고 하는데, 
# 4*3=12이므로 원소의 개수가 맞지 않아서 오류가 난다.
```

```python
array1 = np.arange(10) # -> 0부터 9까지의 정수로 이루어진 1차원 배열 생성
print(array1)

array2 = array1.reshape(-1,5) # -> -1은 자동으로 계산해줌. 열이 5개이므로 행은 2개로 자동 계산
print('array2 shape:',array2.shape) # -> (2, 5)

array3 = array1.reshape(5,-1) # -> 행이 5개이므로 열은 2개로 자동 계산
print('array3 shape:',array3.shape) # -> (5, 2)
```

```python
array1 = np.arange(10) # -> 10개의 원소
array4 = array1.reshape(-1,4) # -> 열이 4개인데 10은 4로 나누어떨어지지 않으므로 에러 발생!
```

```python
array1 = np.arange(8) # -> 0부터 7까지의 정수로 이루어진 1차원 배열 생성 (8개의 원소)
array3d = array1.reshape((2,2,2)) # -> 8개의 원소를 2x2x2 3차원 배열로 변환
print('array3d:\n',array3d.tolist()) # tolist() 메서드로 리스트 형태로 출력

# 3차원 ndarray를 2차원 ndarray로 변환
array5 = array3d.reshape(-1,1) # -> 3차원 배열을 2차원 배열로 변환. 열이 1개이므로 행은 8개로 자동 계산
print('array5:\n',array5.tolist())
print('array5 shape:',array5.shape) # -> (8, 1)

# 1차원 ndarray를 2차원 ndarray로 변환
array6 = array1.reshape(-1,1) # -> 1차원 배열을 2차원 배열로 변환. 열이 1개이므로 행은 8개로 자동 계산
print('array6:\n',array6.tolist())
print('array6 shape:',array6.shape) # -> (8, 1)
```

```python
array1d = np.arange(start=1, stop=10) # -> 1부터 9까지의 정수로 이루어진 1차원 배열 생성
array2d = array1d.reshape(3,3) # -> 3행 3열의 2차원 배열로 변환
print('array2d:\n',array2d)

print('array2d[0:2, 0:2] \n', array2d[0:2, 0:2]) # -> 0~1행, 0~1열 추출 (2x2 배열)
print('array2d[1:3, 0:3] \n', array2d[1:3, 0:3]) # -> 1~2행, 0~2열 추출 (2x3 배열)
print('array2d[1:3, :] \n', array2d[1:3, :]) # -> 1~2행, 전체 열 추출 (2x3 배열)
print('array2d[:, :] \n', array2d[:, :]) # -> 전체 행, 전체 열 추출 (전체 배열)
print('array2d[:2, 1:] \n', array2d[:2, 1:]) # -> 0~1행, 1~2열 추출 (2x2 배열)
print('array2d[:2, 0] \n', array2d[:2, 0]) # -> 0~1행, 0열만 추출 (1차원 배열로 반환됨)
```

```python
array1d = np.arange(start=1, stop=10) # -> 1부터 9까지의 정수로 이루어진 1차원 배열 생성
array2d = array1d.reshape(3,3) # -> 3행 3열의 2차원 배열로 변환

array3 = array2d[[0,1], 2] # -> 팬시 인덱싱: 0행과 1행의 2열 값만 추출 (1차원 배열)
print('array2d[[0,1], 2] => ',array3.tolist()) # -> [3, 6]

array4 = array2d[[0,1], 0:2] # -> 0행과 1행의 0~1열 추출 (2x2 배열)
print('array2d[[0,1], 0:2] => ',array4.tolist()) # -> [[1, 2], [4, 5]]

array5 = array2d[[0,1]] # -> 0행과 1행 전체를 추출 (2x3 배열)
print('array2d[[0,1]] => ',array5.tolist()) # -> [[1, 2, 3], [4, 5, 6]]
```

## 행렬의 정렬, sort( )와 argsort( )

```python
org_array = np.array([ 3, 1, 9, 5]) # <= 정렬되지 않은 배열 생성
print('원본 행렬:', org_array)
# np.sort( )로 정렬 
sort_array1 = np.sort(org_array) # -> np.sort()는 원본을 유지하고 정렬된 배열을 반환
print ('np.sort( ) 호출 후 반환된 정렬 행렬:', sort_array1) # -> [1 3 5 9] 오름차순 정렬됨
print('np.sort( ) 호출 후 원본 행렬:', org_array) # -> [3 1 9 5] 원본은 변경되지 않음
# ndarray.sort( )로 정렬
sort_array2 = org_array.sort() # -> ndarray.sort()는 원본 자체를 정렬하고 None을 반환
print('org_array.sort( ) 호출 후 반환된 행렬:', sort_array2) # -> None
print('org_array.sort( ) 호출 후 원본 행렬:', org_array) # -> [1 3 5 9] 원본이 정렬됨
```

```python
sort_array1_desc = np.sort(org_array)[::-1] # -> 오름차순 정렬 후 [::-1]로 역순 정렬 (내림차순)
print ('내림차순으로 정렬:', sort_array1_desc) # -> [9 5 3 1]
```

```python
array2d = np.array([[8, 12], 
                   [7, 1 ]]) # <= 2행 2열의 2차원 배열 생성

sort_array2d_axis0 = np.sort(array2d, axis=0) # -> axis=0은 행 방향(세로)으로 정렬. 각 열마다 정렬됨
print('로우 방향으로 정렬:\n', sort_array2d_axis0) # -> [[7, 1], [8, 12]]

sort_array2d_axis1 = np.sort(array2d, axis=1) # -> axis=1은 열 방향(가로)으로 정렬. 각 행마다 정렬됨
print('컬럼 방향으로 정렬:\n', sort_array2d_axis1) # -> [[8, 12], [1, 7]]
```

```python
org_array = np.array([ 3, 1, 9, 5]) # <= 정렬되지 않은 배열
sort_indices = np.argsort(org_array) # -> 정렬된 배열의 원본 인덱스를 반환 (값이 아닌 인덱스를 반환)
print(type(sort_indices)) # -> numpy.ndarray
print('행렬 정렬 시 원본 행렬의 인덱스:', sort_indices) # -> [1 0 3 2] (1→3→5→9 순서의 원본 인덱스)
```

```python
org_array = np.array([ 3, 1, 9, 5]) # <= 정렬되지 않은 배열
sort_indices_desc = np.argsort(org_array)[::-1] # -> argsort() 결과를 [::-1]로 역순 정렬 (내림차순)
print('행렬 내림차순 정렬 시 원본 행렬의 인덱스:', sort_indices_desc) # -> [2 3 0 1] (9→5→3→1 순서의 원본 인덱스)
```

```python
import numpy as np

name_array = np.array(['John', 'Mike', 'Sarah', 'Kate', 'Samuel']) # <= 이름 배열
score_array= np.array([78, 95, 84, 98, 88]) # <= 점수 배열

sort_indices_asc = np.argsort(score_array) # -> 점수를 오름차순 정렬했을 때의 인덱스
print('성적 오름차순 정렬 시 score_array의 인덱스:', sort_indices_asc) # -> [0 2 4 1 3] (78→84→88→95→98)
print('성적 오름차순으로 name_array의 이름 출력:', name_array[sort_indices_asc]) # -> 팬시 인덱싱으로 이름을 성적순으로 출력
```

## 선형대수 연산 – 행렬 내적과 전치 행렬 구하기

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]]) # -> 2행 3열의 행렬 A
B = np.array([[7, 8],
              [9, 10],
              [11, 12]]) # -> 3행 2열의 행렬 B

dot_product = np.dot(A, B) # -> 행렬의 내적(곱셈) 계산. A(2x3) * B(3x2) = 결과(2x2)
print('행렬 내적 결과:\n', dot_product) # -> [[58, 64], [139, 154]]
```

```python
A = np.array([[1, 2],
              [3, 4]]) # -> 2행 2열의 행렬 A
transpose_mat = np.transpose(A) # -> 전치 행렬: 행과 열을 바꾼 행렬 (A의 i행 j열 -> 전치의 j행 i열)
print('A의 전치 행렬:\n', transpose_mat) # -> [[1, 3], [2, 4]]
```

```python
lambda_square = lambda x : x ** 2 # -> lambda 식으로 동일한 제곱 함수 정의. 한 줄로 간결하게 표현
print('3의 제곱은:',lambda_square(3)) # -> lambda 함수 호출하여 3의 제곱 계산
```

```python
# 나이에 따라 세분화된 분류를 수행하는 함수 생성. 
def get_category(age): # -> 나이를 받아서 7단계로 세분화하는 함수
    cat = ''
    if age <= 5: cat = 'Baby' # -> 5세 이하: Baby
    elif age <= 12: cat = 'Child' # -> 12세 이하: Child
    elif age <= 18: cat = 'Teenager' # -> 18세 이하: Teenager
    elif age <= 25: cat = 'Student' # -> 25세 이하: Student
    elif age <= 35: cat = 'Young Adult' # -> 35세 이하: Young Adult
    elif age <= 60: cat = 'Adult' # -> 60세 이하: Adult
    else : cat = 'Elderly' # -> 60세 초과: Elderly
    
    return cat

# lambda 식에 위에서 생성한 get_category( ) 함수를 반환값으로 지정. 
# get_category(X)는 입력값으로 'Age' 컬럼 값을 받아서 해당하는 cat 반환
titanic_df['Age_cat'] = titanic_df['Age'].apply(lambda x : get_category(x)) # -> 각 나이를 7단계 카테고리로 분류
titanic_df[['Age','Age_cat']].head() # -> 나이와 카테고리 컬럼만 추출하여 처음 5개 출력
    
```
