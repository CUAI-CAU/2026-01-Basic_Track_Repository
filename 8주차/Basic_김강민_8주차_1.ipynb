## 4.5 GBM(Gradient Boosting Machine)

```python
# GradientBoostingClassifier는 GBM 분류 모델을 구현하기 위한 사이킷런의 클래스이다.
from sklearn.ensemble import GradientBoostingClassifier

# accuracy_score는 실제값과 예측값을 비교하여 정확도를 계산할 때 사용한다.
from sklearn.metrics import accuracy_score

# time 모듈은 GBM 모델 학습과 예측에 걸리는 시간을 측정하기 위해 사용한다.
import time

# warnings 모듈은 실행 중 발생하는 경고 메시지를 숨기기 위해 사용한다.
import warnings
warnings.filterwarnings('ignore')

# 앞에서 정의한 get_human_dataset() 함수를 이용하여
# Human Activity Recognition 데이터셋의 학습 데이터와 테스트 데이터를 불러온다.
X_train, X_test, y_train, y_test = get_human_dataset()

# GBM 모델의 수행 시간을 측정하기 위해 시작 시간을 저장한다.
start_time = time.time()

# GradientBoostingClassifier 객체를 생성한다.
# random_state=0으로 설정하여 실행할 때마다 동일한 결과가 나오도록 한다.
gb_clf = GradientBoostingClassifier(random_state=0)

# 학습 데이터를 이용하여 GBM 모델을 학습시킨다.
gb_clf.fit(X_train, y_train)

# 학습된 GBM 모델을 이용하여 테스트 데이터의 클래스를 예측한다.
gb_pred = gb_clf.predict(X_test)

# 실제 테스트 정답값 y_test와 예측값 gb_pred를 비교하여 정확도를 계산한다.
gb_accuracy = accuracy_score(y_test, gb_pred)

# GBM 모델의 테스트 정확도를 출력한다.
print('GBM 정확도: {0:.4f}'.format(gb_accuracy))

# 현재 시간에서 시작 시간을 빼서 GBM 수행 시간을 출력한다.
print("GBM 수행 시간: {0:.1f} 초 ".format(time.time() - start_time))
```

```python
# GridSearchCV는 여러 하이퍼파라미터 조합을 비교하여
# 가장 성능이 좋은 조합을 찾기 위해 사용한다.
from sklearn.model_selection import GridSearchCV

# GBM에서 조정할 하이퍼파라미터 후보들을 설정한다.
# n_estimators는 생성할 약한 학습기의 개수이다.
# learning_rate는 각 약한 학습기가 전체 모델에 반영되는 정도를 의미한다.
params = {
    'n_estimators': [100, 500],
    'learning_rate': [0.05, 0.1]
}

# GridSearchCV 객체를 생성한다.
# gb_clf 모델을 기반으로 params에 설정한 하이퍼파라미터 조합을 탐색한다.
# cv=2는 데이터를 2개의 폴드로 나누어 교차 검증을 수행한다는 의미이다.
# verbose=1은 학습 진행 상황을 출력하도록 설정한 것이다.
grid_cv = GridSearchCV(gb_clf, param_grid=params, cv=2, verbose=1)

# 학습 데이터를 이용하여 GridSearchCV를 수행한다.
# 즉, 여러 하이퍼파라미터 조합 중 가장 좋은 조합을 찾는다.
grid_cv.fit(X_train, y_train)

# GridSearchCV를 통해 찾은 최적의 하이퍼파라미터를 출력한다.
print('최적 하이퍼 파라미터:\n', grid_cv.best_params_)

# 교차 검증에서 얻은 최고 예측 정확도를 출력한다.
print('최고 예측 정확도: {0:.4f}'.format(grid_cv.best_score_))
```

```python
# GridSearchCV를 통해 최적으로 학습된 estimator를 이용하여
# 테스트 데이터에 대한 예측을 수행한다.
gb_pred = grid_cv.best_estimator_.predict(X_test)

# 실제 테스트 정답값과 최적 GBM 모델의 예측값을 비교하여 정확도를 계산한다.
gb_accuracy = accuracy_score(y_test, gb_pred)

# 최적 하이퍼파라미터가 적용된 GBM 모델의 테스트 정확도를 출력한다.
print('GBM 정확도: {0:.4f}'.format(gb_accuracy))
```
